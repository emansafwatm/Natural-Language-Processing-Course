# 1. Introduction

Natural Language Processing (NLP) is the field concerned with enabling computers to process, analyze, interpret, and generate human language.

NLP supports many technologies used every day:

- web search;
- machine translation;
- spam filtering;
- text classification;
- sentiment analysis;
- virtual assistants;
- automatic summarization;
- question answering;
- conversational systems;
- Large Language Models.

Human language is not a formal system designed for machines. It is flexible, ambiguous, culturally influenced, context-sensitive, and continuously changing. These properties make language expressive for humans but difficult for computers.

NLP addresses this difficulty by combining ideas from linguistics, computer science, mathematics and statistics, and Artificial Intelligence.

This lesson introduces the conceptual foundations required for the remainder of the course.

> **Key Idea**
>
> NLP is not a single algorithm or model. It is a broad field containing many tasks, representations, methods, datasets, and evaluation strategies.

# 2. What Is Natural Language?

A **natural language** is a language that developed through human communication rather than being formally designed for a machine.

Examples include Arabic, English, French, Chinese, Spanish, Japanese, and German. Natural languages can appear in spoken, written, or signed form.

## Natural Language and Programming Language

| Natural Language | Programming Language |
|---|---|
| Developed through human use | Deliberately designed |
| Often ambiguous | Intended to be precise |
| Depends heavily on context | Governed by formal syntax |
| Tolerates incomplete or imperfect expressions | Syntax errors often prevent execution |
| Changes continuously | Changes through controlled specifications |

Consider the sentence:

> I saw the man with the telescope.

This sentence has at least two interpretations: the speaker used a telescope to see the man, or the man being observed carried a telescope. A human reader may resolve the ambiguity using context. A computer requires linguistic information, contextual evidence, or a trained model.

## Core Properties of Natural Language

### Ambiguity

A word, phrase, or sentence may support multiple interpretations.

> The bank is closed.

The word *bank* may refer to a financial institution or the side of a river.

### Context Dependence

Meaning often depends on surrounding text, shared knowledge, or the communication situation.

> It is cold here.

This may be an observation, a complaint, or an indirect request to close a window.

### Variability

The same meaning can be expressed in different forms:

- The meeting starts at nine.
- The meeting begins at nine.
- The meeting will commence at nine.

### Creativity

Speakers continually create new expressions, abbreviations, slang, and domain-specific terminology.

### Multilingual Diversity

Languages differ in writing systems, word order, morphology, punctuation, agreement patterns, word segmentation, and available digital resources.

# 3. What Is Natural Language Processing?

Natural Language Processing is an interdisciplinary field that develops computational methods for working with human language.

An NLP system may receive language as input and produce a label, score, structured representation, extracted fact, generated response, translation, summary, or prediction.

| Input | NLP Task | Output |
|---|---|---|
| Customer review | Sentiment analysis | Positive, negative, or neutral |
| News article | Summarization | Shorter representation |
| Arabic sentence | Machine translation | English sentence |
| Email | Spam classification | Spam or not spam |
| Question and passage | Question answering | Answer span |
| Unstructured document | Information extraction | Names, dates, organizations |
| User prompt | Text generation | Generated continuation or response |

NLP therefore includes both **language understanding** and **language generation**.

## Understanding and Generation

### Natural Language Understanding

Natural Language Understanding focuses on interpreting language. Typical objectives include identifying intent, recognizing entities, detecting sentiment, assigning labels, extracting relationships, resolving references, and representing meaning.

### Natural Language Generation

Natural Language Generation focuses on producing language. Typical objectives include generating answers, summarizing documents, translating text, producing descriptions, rewriting content, and creating dialogue responses.

Modern systems often combine understanding and generation within one workflow.

# 4. NLP and Related AI Fields

NLP is closely connected to Artificial Intelligence, Machine Learning, Deep Learning, and Large Language Models, but these terms are not interchangeable.

- **Artificial Intelligence** is the broad field concerned with building systems that perform tasks associated with human intelligence.
- **Machine Learning** is a subfield of AI in which models learn patterns from data rather than relying only on manually written rules.
- **Deep Learning** is a subfield of Machine Learning based on multilayer neural networks.
- **Large Language Models** are neural language models, usually based on Transformer architectures, trained on large text collections and designed to perform or support many language tasks.

NLP is a **problem domain** centered on language. Machine Learning and Deep Learning are **methodological families** that can be used to solve NLP problems.

A rule-based tokenizer is an NLP system without Machine Learning. A neural sentiment classifier is both an NLP system and a Deep Learning system. A general-purpose LLM is a modern model used for many NLP tasks.

```text
Artificial Intelligence
├── Rule-based systems
├── Machine Learning
│   └── Deep Learning
│       └── Transformer-based models
│           └── Large Language Models
└── Other AI approaches

Natural Language Processing
├── Rule-based NLP
├── Statistical NLP
├── Machine-learning NLP
├── Deep-learning NLP
└── LLM-based NLP
```

The two hierarchies overlap because NLP defines the language-related problems, while AI and Machine Learning define possible solution approaches.

# 5. Why Human Language Is Difficult

Human language contains many forms of uncertainty and variation.

## 5.1 Lexical Ambiguity
A single word may have several meanings.

> She went to the bank.

## 5.2 Syntactic Ambiguity
A sentence may have more than one grammatical structure.

> I saw the student with the camera.

## 5.3 Semantic Ambiguity
A grammatically clear sentence may still support multiple meanings.

> Visiting relatives can be tiring.

## 5.4 Pragmatics
The intended meaning may differ from the literal wording.

> Could you open the door?

## 5.5 Coreference
Pronouns and noun phrases may refer to earlier entities.

> Sara gave Mona the report because she had finished it.

## 5.6 Idioms and Figurative Language

> The project is still up in the air.

## 5.7 Sarcasm

> Wonderful. The server crashed again.

## 5.8 Domain Variation

The word *cell* has different meanings in biology, telecommunications, spreadsheets, and prisons.

## 5.9 Multilingual and Morphological Complexity

Languages differ substantially in structure. Arabic, for example, uses rich morphology and productive patterns that can encode substantial information within a single word.

> **Key Takeaways**
>
> Language processing is difficult because meaning depends on words, structure, context, speaker intention, domain, culture, and prior knowledge. A successful NLP system must model at least part of this information.

# 8. A Typical NLP Workflow

```text
Problem Definition
        ↓
Data Collection
        ↓
Data Inspection and Cleaning
        ↓
Text Representation
        ↓
Model Selection or Rule Design
        ↓
Training or Configuration
        ↓
Evaluation
        ↓
Deployment
        ↓
Monitoring and Improvement
```

## Problem Definition
The task, target users, constraints, risks, and evaluation criteria must be defined before implementation.

## Data Collection
Data may come from documents, databases, websites, transcripts, logs, surveys, or public datasets.

## Data Inspection and Cleaning
The developer examines encoding, missing values, duplicates, noise, labels, imbalance, and privacy concerns.

## Text Representation
Text must be converted into a form that an algorithm can process. Representations may include counts, TF-IDF vectors, static embeddings, contextual embeddings, or token sequences.

## Model or Rule Selection
The solution may use linguistic rules, statistical methods, classical Machine Learning, neural networks, pretrained Transformer models, or Large Language Models.

## Evaluation
Performance must be measured using task-appropriate metrics and qualitative analysis.

## Deployment and Monitoring
Production systems require monitoring for errors, drift, latency, fairness, safety, and changing language use.

# 9. First Python Text Analysis

The first examples use Python's standard library and Matplotlib. No external language model is required.

````python
# Standard library imports
import re
from collections import Counter
from typing import List

# Third-party imports
import matplotlib.pyplot as plt
````

## 9.1 Define a Sample Text

````python
text = (
    "Natural Language Processing enables computers to work with human language. "
    "NLP systems can classify documents, translate text, answer questions, "
    "and generate useful summaries."
)

print(text)
````

## 9.2 Count Characters, Words, and Sentences

A character count includes letters, spaces, and punctuation. A simple word count can be produced by splitting on whitespace. Sentence counting is more difficult because punctuation does not always indicate a sentence boundary.

````python
character_count = len(text)
simple_words = text.split()
word_count = len(simple_words)

sentences = [
    sentence.strip()
    for sentence in re.split(r"[.!?]+", text)
    if sentence.strip()
]
sentence_count = len(sentences)

print(f"Characters: {character_count}")
print(f"Words: {word_count}")
print(f"Sentences: {sentence_count}")
````

## 9.3 Basic Tokenization

Tokenization divides text into smaller units called **tokens**. The following regular expression extracts alphabetic word tokens. It is intentionally simple and will not handle every language or punctuation pattern correctly.

````python
def tokenize_words(value: str) -> List[str]:
    """Return lowercase alphabetic word tokens from text."""
    return re.findall(r"[A-Za-z]+(?:'[A-Za-z]+)?", value.lower())


tokens = tokenize_words(text)

print(tokens)
print(f"Token count: {len(tokens)}")
````

## 9.4 Word Frequencies

````python
frequency = Counter(tokens)

for word, count in frequency.most_common():
    print(f"{word:<12} {count}")
````

## 9.5 Visualize the Most Frequent Words

````python
top_items = frequency.most_common(10)
words = [item[0] for item in top_items]
counts = [item[1] for item in top_items]

plt.figure(figsize=(10, 5))
plt.bar(words, counts)
plt.title("Most Frequent Words in the Sample Text")
plt.xlabel("Word")
plt.ylabel("Frequency")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

## 9.6 Remove a Small Stop-Word Set

Common function words such as *the*, *and*, and *to* are sometimes removed in classical NLP workflows. This is not always appropriate. In many modern systems, stop words are retained because they contribute to syntax and meaning.

````python
stop_words = {
    "a", "an", "and", "are", "as", "at", "be", "by", "for", "from",
    "in", "is", "it", "of", "on", "or", "that", "the", "to", "with"
}

content_tokens = [token for token in tokens if token not in stop_words]

print(content_tokens)
````

## Interpretation

The code above illustrates several important ideas:

1. text must be segmented before many forms of analysis;
2. preprocessing choices affect the resulting tokens;
3. simple rules are useful for demonstrations but may fail on realistic language;
4. token frequencies provide a basic representation of textual content;
5. preprocessing decisions must depend on the task rather than habit.

# 10. Mini Case Study

## Comparing Three Short Reviews

The goal is to compare vocabulary and detect simple opinion words without using a trained sentiment model.

This example is intentionally rule-based. It demonstrates the structure of a small NLP workflow:

1. define the data;
2. tokenize the text;
3. count selected indicators;
4. produce a simple score;
5. interpret the result;
6. discuss limitations.

````python
reviews = {
    "review_1": "The course is clear, practical, and very useful.",
    "review_2": "The explanation is confusing and the examples are weak.",
    "review_3": "The content is detailed, but some sections are difficult."
}

positive_words = {"clear", "practical", "useful", "detailed", "strong", "good"}
negative_words = {"confusing", "weak", "difficult", "bad", "unclear"}


def simple_sentiment_score(value: str) -> int:
    """Return positive-word count minus negative-word count."""
    review_tokens = tokenize_words(value)
    positive_count = sum(token in positive_words for token in review_tokens)
    negative_count = sum(token in negative_words for token in review_tokens)
    return positive_count - negative_count


for review_id, review_text in reviews.items():
    score = simple_sentiment_score(review_text)
    print(f"{review_id}: score={score:+d} | {review_text}")
````

## Case Study Interpretation

The rule assigns a positive score when positive terms outnumber negative terms, a negative score when negative terms outnumber positive terms, and zero when the counts are balanced.

This method is easy to explain but has major limitations:

- it does not understand negation;
- it ignores word order;
- it does not handle sarcasm;
- it depends on a manually created vocabulary;
- it treats all words as equally important;
- it may fail across domains and languages;
- it cannot represent context.

For example, a simple word list may count *useful* as positive in the sentence *The course is not useful*, even though the sentence is negative. This limitation motivates more advanced representations and models introduced later in the course.

````python
difficult_examples = [
    "The course is not useful.",
    "The examples are hardly clear.",
    "Wonderful, another confusing explanation."
]

for example in difficult_examples:
    score = simple_sentiment_score(example)
    print(f"score={score:+d} | {example}")
````

# 11. Common Misconceptions

## Misconception 1: NLP Means Chatbots
Chatbots are one application of NLP. The field also includes classification, extraction, search, translation, summarization, and many other tasks.

## Misconception 2: Every NLP System Uses Deep Learning
Rule-based and classical statistical methods remain useful, especially when data is limited, transparency is important, or the task is narrow.

## Misconception 3: A High Accuracy Score Guarantees a Good System
A model may perform well on one dataset and fail in a new domain. Evaluation must consider data quality, class imbalance, robustness, fairness, latency, and user impact.

## Misconception 4: Tokenization Is Trivial
Tokenization depends on language, script, model architecture, and task. A method that works for English may not work well for Arabic, Chinese, or code-mixed text.

## Misconception 5: Large Language Models Understand Language Exactly Like Humans
LLMs can produce highly capable language behavior, but their internal mechanisms, limitations, grounding, reliability, and reasoning differ from human cognition.
