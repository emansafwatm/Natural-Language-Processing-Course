## Introduction
Natural Language Processing has evolved over more than seventy years. Each generation emerged to address limitations of the previous one, leading from handcrafted linguistic rules to modern foundation models.

## Why Study NLP History?
Understanding the history of NLP helps explain why different approaches exist, their assumptions, strengths, and limitations.

## Major Eras of NLP

| Period | Dominant Paradigm |
|---|---|
|1950s–1980s|Rule-Based NLP|
|Late 1980s–1990s|Statistical NLP|
|Late 1990s–2010s|Machine Learning|
|2013–2017|Deep Learning|
|2017–Present|Transformer Era|
|2020–Present|Large Language Models|

## The Rule-Based Era (1950s–1980s)
Early NLP systems relied on handcrafted linguistic rules, lexicons, and grammars. They were interpretable but difficult to scale and maintain.

```
# Example of simple rule-based logic
def classify(sentence):
    if '?' in sentence:
        return 'Question'
    if sentence.startswith('Please'):
        return 'Request'
    return 'Statement'

for s in ['How are you?', 'Please open the door.', 'NLP is fascinating.']:
    print(s, '->', classify(s))
```
