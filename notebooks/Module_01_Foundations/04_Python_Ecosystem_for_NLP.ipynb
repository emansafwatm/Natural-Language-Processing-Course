## 2.1 Installation Name Versus Import Name

The package name used with `pip` is not always identical to the name used
inside Python.

| Install with pip | Import in Python |
|---|---|
| `scikit-learn` | `import sklearn` |
| `beautifulsoup4` | `from bs4 import BeautifulSoup` |
| `pillow` | `from PIL import Image` |
| `pyyaml` | `import yaml` |
| `opencv-python` | `import cv2` |

This distinction explains why `pip install sklearn` fails: the maintained
installation package is `scikit-learn`, while the import namespace remains
`sklearn`.

# 3. Jupyter Notebook Practices

Jupyter notebooks combine executable code, output, equations, explanations,
and visualizations.

Important practices:

- run cells from top to bottom before publishing;
- restart the kernel and run all cells to detect hidden state;
- place imports and configuration near the beginning;
- avoid redefining the same variable unpredictably;
- keep each code cell focused on one concept;
- explain why code is being used, not only what it prints;
- avoid storing secrets, tokens, private paths, or personal data;
- clear unnecessary output before committing very large notebooks;
- record package versions when reproducibility matters.

A notebook that works only because cells were executed in a private order is
not reproducible.

# 4. Python Text and Unicode

Python 3 strings represent Unicode text. This is essential for multilingual
NLP because language data may contain Latin, Arabic, Chinese, accented
characters, emoji, and many other writing systems.

````python
examples = [
    "Natural language processing",
    "معالجة اللغة الطبيعية",
    "traitement automatique du langage",
    "自然语言处理",
    "NLP is useful 😊",
]

for text in examples:
    print(f"{text!r}")
    print("  characters:", len(text))
    print("  UTF-8 bytes:", len(text.encode("utf-8")))
````

Character count and encoded byte count are not necessarily equal. File size
and network transmission operate on bytes, while most Python text operations
work with Unicode characters.

## 4.1 Unicode Normalization

Visually similar text can have different internal Unicode sequences.
Normalization provides a consistent representation before matching or
comparison.

````python
import unicodedata

composed = "café"
decomposed = "cafe\u0301"

print("Equal before normalization:", composed == decomposed)
print("Composed code points:", [hex(ord(ch)) for ch in composed])
print("Decomposed code points:", [hex(ord(ch)) for ch in decomposed])

normalized_composed = unicodedata.normalize("NFC", composed)
normalized_decomposed = unicodedata.normalize("NFC", decomposed)

print("Equal after NFC normalization:",
      normalized_composed == normalized_decomposed)
````

Unicode normalization must be task-aware. It should not be confused with
removing diacritics or changing spelling, both of which may discard meaningful
linguistic information.

## 4.2 Core String Operations

````python
sentence = "Natural Language Processing with Python"

print("Lowercase:", sentence.lower())
print("Starts with 'Natural':", sentence.startswith("Natural"))
print("Contains 'Python':", "Python" in sentence)
print("Words:", sentence.split())
print("Replaced:", sentence.replace("Python", "open-source tools"))
print("First 7 characters:", sentence[:7])
````

# 5. Collections, Functions, and Type Hints

NLP programs frequently use:

- lists for ordered tokens or documents;
- tuples for fixed records or n-grams;
- dictionaries for counts, labels, metadata, and configuration;
- sets for unique vocabulary;
- functions for reusable processing;
- type hints for clearer interfaces.

````python
from collections import Counter

documents = [
    "language models learn patterns",
    "language processing uses models",
    "models require evaluation",
]

tokens = [token for document in documents for token in document.split()]
vocabulary = set(tokens)
frequencies = Counter(tokens)

print("Documents:", len(documents))
print("Tokens:", len(tokens))
print("Vocabulary size:", len(vocabulary))
print("Most common:", frequencies.most_common(5))
````

````python
def normalize_whitespace(text: str) -> str:
    """Collapse repeated whitespace and remove outer whitespace."""
    if not isinstance(text, str):
        raise TypeError("text must be a string")
    return " ".join(text.split())


samples = [
    "  NLP   uses   text  ",
    "One\nmessage\twith several spaces",
]

for sample in samples:
    print(repr(normalize_whitespace(sample)))
````

# 6. Files, Paths, CSV, and JSON

The standard library supports many common data operations without additional
packages.

- `pathlib` manages cross-platform file paths;
- `csv` reads and writes comma-separated data;
- `json` stores structured metadata;
- explicit UTF-8 encoding prevents platform-dependent text errors.

````python
import json
from pathlib import Path
from tempfile import TemporaryDirectory

sample_records = [
    {"id": 1, "text": "The lesson is clear.", "label": "positive"},
    {"id": 2, "text": "The example is confusing.", "label": "negative"},
]

with TemporaryDirectory() as temporary_directory:
    data_path = Path(temporary_directory) / "records.json"

    data_path.write_text(
        json.dumps(sample_records, ensure_ascii=False, indent=2),
        encoding="utf-8",
    )

    loaded_records = json.loads(
        data_path.read_text(encoding="utf-8")
    )

    print("Temporary path:", data_path)
    print("Loaded records:", loaded_records)
````

`ensure_ascii=False` keeps non-ASCII text readable in JSON. The file should
still be opened explicitly with UTF-8 encoding.

# 7. Regular Expressions

Python's `re` module provides pattern-based text matching.

Regular expressions are useful for:

- validating structured identifiers;
- locating email addresses, URLs, dates, or numbers;
- masking predictable sensitive patterns;
- controlled normalization;
- rule-based baselines.

They are not a complete language-understanding method.

````python
import re

message = (
    "Contact support@example.com or visit "
    "https://example.com/help before 2026-08-01."
)

email_pattern = re.compile(
    r"\b[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}\b"
)
url_pattern = re.compile(r"https?://\S+")
date_pattern = re.compile(r"\b\d{4}-\d{2}-\d{2}\b")

print("Emails:", email_pattern.findall(message))
print("URLs:", url_pattern.findall(message))
print("Dates:", date_pattern.findall(message))
````

````python
masked_message = email_pattern.sub("EMAIL_ADDRESS", message)
masked_message = url_pattern.sub("URL", masked_message)

print(masked_message)
````

Use raw strings such as `r"\d+"` for regex patterns. They reduce conflicts
between Python string escaping and regular-expression escaping.

# 8. NumPy for Numerical Data

NumPy provides efficient multidimensional arrays and numerical operations. NLP
models use numerical representations for counts, embeddings, probabilities,
gradients, and attention values.

````python
import numpy as np

document_vectors = np.array(
    [
        [2.0, 1.0, 0.0],
        [1.0, 2.0, 0.0],
        [0.0, 0.0, 3.0],
    ]
)

print("Shape:", document_vectors.shape)
print("Column means:", document_vectors.mean(axis=0))
print("Row sums:", document_vectors.sum(axis=1))
````

````python
def cosine_similarity(vector_a: np.ndarray,
                      vector_b: np.ndarray) -> float:
    denominator = np.linalg.norm(vector_a) * np.linalg.norm(vector_b)
    if denominator == 0:
        return 0.0
    return float(np.dot(vector_a, vector_b) / denominator)


print(
    "Similarity between document 1 and 2:",
    round(cosine_similarity(document_vectors[0],
                            document_vectors[1]), 3),
)
print(
    "Similarity between document 1 and 3:",
    round(cosine_similarity(document_vectors[0],
                            document_vectors[2]), 3),
)
````

# 9. Pandas for Text Datasets

Pandas represents structured datasets as `DataFrame` objects. It is commonly
used for:

- loading CSV, JSON, Excel, and database results;
- inspecting columns and missing values;
- filtering and grouping records;
- creating derived features;
- joining text with labels and metadata;
- preparing data for visualization or modeling.

````python
import pandas as pd

feedback = pd.DataFrame(
    [
        (1, "The explanation is clear and practical.", "positive", "web"),
        (2, "The notebook crashes during installation.", "negative", "email"),
        (3, "Useful examples and good organization.", "positive", "web"),
        (4, "The instructions are difficult to follow.", "negative", "chat"),
        (5, "The lesson is detailed and helpful.", "positive", "email"),
        (6, "Some code cells produce an error.", "negative", "chat"),
        (7, "Clear lesson with practical code.", "positive", "web"),
        (8, "Installation steps are confusing.", "negative", "email"),
    ],
    columns=["id", "text", "label", "channel"],
)

feedback
````

````python
feedback["character_count"] = feedback["text"].str.len()
feedback["word_count"] = feedback["text"].str.split().str.len()

summary = (
    feedback
    .groupby("label")
    .agg(
        examples=("id", "count"),
        mean_words=("word_count", "mean"),
        mean_characters=("character_count", "mean"),
    )
)

summary
````

Vectorized string methods such as `.str.lower()` and `.str.len()` are usually
preferable to manually iterating over each DataFrame row.

# 10. Matplotlib for Visualization

Visualizations help inspect distributions, compare categories, and communicate
results. A chart should have a clear title, labeled axes, and an explicit
interpretation.

````python
import matplotlib.pyplot as plt

label_counts = feedback["label"].value_counts()

plt.figure(figsize=(6, 4))
plt.bar(label_counts.index, label_counts.values)
plt.title("Feedback Label Distribution")
plt.xlabel("Label")
plt.ylabel("Number of examples")
plt.tight_layout()
plt.show()
````

The dataset is deliberately balanced for a compact classroom demonstration.
Real datasets should be inspected for imbalance before selecting metrics or
sampling strategies.

# 11. scikit-learn for Classical NLP

scikit-learn provides:

- preprocessing and feature extraction;
- classical supervised and unsupervised models;
- train-test splitting and cross-validation;
- evaluation metrics;
- reproducible pipelines;
- parameter search.

The library is imported as `sklearn`, even though it is installed with the
package name `scikit-learn`.

````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import LeaveOneOut, cross_val_predict
from sklearn.pipeline import Pipeline

classifier = Pipeline(
    [
        (
            "tfidf",
            TfidfVectorizer(
                lowercase=True,
                ngram_range=(1, 2),
            ),
        ),
        (
            "model",
            LogisticRegression(
                max_iter=1000,
                random_state=lesson_configuration["random_state"],
            ),
        ),
    ]
)

classifier
````

````python
from sklearn.metrics import accuracy_score, classification_report

# Leave-one-out evaluation allows every small example to be tested once.
predicted_labels = cross_val_predict(
    classifier,
    feedback["text"],
    feedback["label"],
    cv=LeaveOneOut(),
)

print("Accuracy:", round(
    accuracy_score(feedback["label"], predicted_labels), 3
))
print()
print(classification_report(
    feedback["label"],
    predicted_labels,
    zero_division=0,
))
````

The dataset is too small to support a meaningful production claim. The code
demonstrates library integration and evaluation mechanics.

````python
classifier.fit(feedback["text"], feedback["label"])

new_feedback = [
    "The examples are useful and easy to understand.",
    "The installation instructions fail on my computer.",
]

for text, label in zip(
    new_feedback,
    classifier.predict(new_feedback),
):
    print(f"{text!r} -> {label}")
````

# 14. Library Selection Guide

Before selecting a library, ask:

1. Is the task linguistic analysis, machine learning, generation, or data
   engineering?
2. Does the task require a pretrained language model?
3. Which languages and writing systems must be supported?
4. Is the priority teaching clarity, research flexibility, or production
   throughput?
5. Is GPU acceleration required?
6. Must the system run offline or on limited hardware?
7. Are model size, license, privacy, and latency acceptable?
8. How will the tool be tested, versioned, and maintained?

Avoid choosing a large framework when the standard library or a compact model
solves the requirement adequately.
