# 1. Language as a Structured System

Human language is not an unordered list of words. It is a layered system in
which small units combine into larger units.

```text
Sounds or Characters
         ↓
      Morphemes
         ↓
       Words
         ↓
  Phrases and Clauses
         ↓
     Sentences
         ↓
  Discourse and Dialogue
```

Meaning depends on several interacting sources:

- the forms that appear;
- the internal structure of words;
- the order and grammatical relations among words;
- lexical and compositional meaning;
- the situation and speaker intention;
- earlier sentences and shared knowledge.

> **Key Idea**
>
> Many NLP errors occur because a system models one linguistic level while
> ignoring another level required by the task.

# 2. Linguistic Units and Levels

| Level | Main question | Example NLP tasks |
|---|---|---|
| Phonetics and phonology | How is language sounded and organized? | Speech recognition, speech synthesis |
| Orthography | How is language written? | Normalization, OCR, spelling correction |
| Morphology | How are words internally structured? | Lemmatization, stemming, morphological analysis |
| Syntax | How do words combine grammatically? | Parsing, grammar checking, extraction |
| Semantics | What do expressions mean? | Similarity, entailment, word-sense disambiguation |
| Pragmatics | What does the speaker intend in context? | Dialogue, intent recognition, irony detection |
| Discourse | How do sentences connect? | Summarization, coreference, coherence analysis |

These levels overlap. Named Entity Recognition, machine translation, and
question answering often require information from several levels at once.

````python
import pandas as pd

linguistic_levels = pd.DataFrame(
    [
        ("Orthography", "written forms", "normalization and spelling"),
        ("Morphology", "word structure", "lemmatization and analysis"),
        ("Syntax", "sentence structure", "parsing and grammatical relations"),
        ("Semantics", "literal meaning", "similarity and entailment"),
        ("Pragmatics", "meaning in context", "intent and dialogue"),
        ("Discourse", "cross-sentence structure", "coreference and summarization"),
    ],
    columns=["Level", "Primary focus", "Example NLP use"],
)

linguistic_levels
````

# 3. Sound, Writing, and Orthography

Spoken and written language are related but not identical.

Written NLP must handle:

- capitalization;
- punctuation;
- spelling variation;
- diacritics;
- scripts and writing direction;
- whitespace conventions;
- abbreviations;
- emojis and symbols;
- Unicode representation.

The same visible text may have different internal Unicode sequences, and the
same linguistic expression may be written differently across communities or
platforms.

````python
examples = [
    "NLP",
    "Natural-language processing",
    "natural language processing",
    "معالجة اللغة الطبيعية",
    "معالَجة اللُّغة الطبيعيّة",
    "NLP 😊",
]

for item in examples:
    print(f"{item!r:<35} characters={len(item):>2} bytes={len(item.encode('utf-8')):>2}")
````

## 3.1 Normalization Versus Information Loss

Normalization can make equivalent forms easier to compare, but aggressive
normalization may remove useful evidence.

For example:

- lowercasing removes capitalization cues;
- punctuation removal may erase sentence boundaries or emotion;
- Arabic diacritic removal may merge forms that are linguistically distinct;
- emoji removal may discard sentiment or intent information.

The correct transformation depends on the task.

````python
import unicodedata

text = "cafe\u0301"
normalized = unicodedata.normalize("NFC", text)

print("Original code points:  ", [hex(ord(character)) for character in text])
print("Normalized code points:", [hex(ord(character)) for character in normalized])
print("Visible original:", text)
print("Visible normalized:", normalized)
````

# 4. Morphology

Morphology studies the internal structure of words.

A **morpheme** is a minimal unit that contributes meaning or grammatical
function.

Examples:

```text
un + help + ful
walk + ed
cat + s
re + write
```

Morphemes may express:

- lexical meaning;
- number;
- tense;
- person;
- gender;
- definiteness;
- negation;
- derivation.

## 4.1 Inflection and Derivation

**Inflection** changes grammatical form without creating a fundamentally new
lexeme.

```text
walk → walked
cat → cats
```

**Derivation** creates a new word or changes lexical category.

```text
happy → happiness
teach → teacher
```

The distinction is useful in lemmatization, search, information retrieval,
machine translation, and language modeling.

````python
morphological_examples = pd.DataFrame(
    [
        ("cats", "cat + plural", "inflection"),
        ("walked", "walk + past", "inflection"),
        ("teacher", "teach + agent noun", "derivation"),
        ("unhelpful", "negative + help + adjective", "derivation"),
    ],
    columns=["Word", "Illustrative analysis", "Process"],
)

morphological_examples
````

## 4.2 Morphological Typology

Languages distribute grammatical information differently.

- **Analytic languages** often use separate words and word order.
- **Agglutinative languages** may combine several relatively clear affixes.
- **Fusional languages** may express several grammatical features in one form.
- **Root-and-pattern morphology**, found prominently in Arabic and other
  Semitic languages, combines roots with vocalic and structural patterns.

These differences affect tokenization, vocabulary size, data sparsity,
lemmatization, and translation.

# 5. Syntax

Syntax studies how words combine into phrases, clauses, and sentences.

Consider:

```text
The researcher analyzed the document.
```

A simplified structure is:

```text
Sentence
├── Noun Phrase: The researcher
└── Verb Phrase
    ├── Verb: analyzed
    └── Noun Phrase: the document
```

Word order alone is not always sufficient. Grammatical relations such as
subject, object, modifier, and complement help determine interpretation.

## 5.1 Constituency and Dependency

**Constituency syntax** groups words into nested phrases.

**Dependency syntax** represents directed relations among words.

```text
analyzed
├── subject → researcher
└── object  → document
```

Constituency is useful for phrase structure. Dependency analysis is often
useful for relation extraction, information extraction, and grammar-aware
features.

````python
dependency_relations = [
    ("analyzed", "researcher", "subject"),
    ("researcher", "The", "determiner"),
    ("analyzed", "document", "object"),
    ("document", "the", "determiner"),
]

pd.DataFrame(
    dependency_relations,
    columns=["Head", "Dependent", "Relation"],
)
````

## 5.2 Agreement and Long-Distance Relations

Syntax may require relations across several words.

```text
The results of the experiment are promising.
```

The verb agrees with the plural head noun *results*, not with the nearby
singular noun *experiment*. Such dependencies challenge systems that rely
only on short local patterns.

# 6. Semantics

Semantics studies conventional meaning.

Important distinctions include:

- lexical meaning of individual words;
- compositional meaning of phrases and sentences;
- semantic roles such as agent, patient, and instrument;
- synonymy and antonymy;
- hyponymy and category relations;
- ambiguity and word senses;
- entailment and contradiction.

## 6.1 Compositionality

Sentence meaning is influenced by both word meanings and structure.

```text
The dog chased the cat.
The cat chased the dog.
```

The two sentences contain the same main words, but their grammatical
relations produce different meanings. A Bag-of-Words representation cannot
fully capture this distinction.

````python
from collections import Counter

sentences = [
    "the dog chased the cat",
    "the cat chased the dog",
]

for sentence in sentences:
    print(Counter(sentence.split()))

print(
    "\nThe word-count representations are identical, "
    "although the sentence meanings differ."
)
````

## 6.2 Semantic Roles

Semantic-role analysis identifies who did what to whom, when, where, and how.

In:

> Sara opened the door with a key.

- `Sara` is the agent;
- `the door` is the affected entity;
- `a key` is the instrument.

Semantic roles support information extraction, question answering, and event
understanding.

# 7. Pragmatics

Pragmatics studies interpretation in context, including speaker intention and
shared knowledge.

Consider:

> Can you open the window?

Literally, the sentence asks about ability. In ordinary conversation, it is
usually interpreted as a request.

Pragmatic interpretation may depend on:

- the speaker and listener;
- location and time;
- social relationship;
- shared assumptions;
- politeness;
- indirect speech acts;
- irony or sarcasm.

````python
utterances = pd.DataFrame(
    [
        ("Can you pass the salt?", "question about ability", "request"),
        ("It is cold in here.", "temperature statement", "possible request to close a window"),
        ("Great job!", "positive evaluation", "praise or sarcasm, depending on context"),
    ],
    columns=["Utterance", "Literal form", "Possible contextual function"],
)

utterances
````

# 8. Discourse

Discourse concerns relationships across sentences and conversational turns.

Important phenomena include:

- reference and coreference;
- coherence;
- topic continuity;
- discourse relations such as cause, contrast, and elaboration;
- turn-taking;
- dialogue state;
- information structure.

Example:

```text
Mona submitted the paper.
She received confirmation the next day.
```

Interpreting *She* requires linking it to *Mona*.

````python
discourse = [
    ("Mona", "submitted", "the paper"),
    ("She", "received", "confirmation"),
]

print("Sentence 1:", discourse[0])
print("Sentence 2:", discourse[1])
print("Coreference hypothesis: 'She' refers to 'Mona'.")
````

# 9. Ambiguity

Language permits multiple interpretations.

## 9.1 Lexical Ambiguity

> The bank is nearby.

*Bank* may refer to a financial institution or the side of a river.

## 9.2 Syntactic Ambiguity

> I saw the researcher with the telescope.

The telescope may be the viewing instrument or an object associated with the
researcher.

## 9.3 Semantic Ambiguity

> Every student read a book.

This may describe one shared book or a potentially different book for each
student.

## 9.4 Pragmatic Ambiguity

> That was impressive.

Depending on context and tone, it may express genuine praise or irony.

````python
ambiguity_examples = pd.DataFrame(
    [
        ("bank", "lexical", "financial institution or river side"),
        ("I saw the researcher with the telescope.", "syntactic", "two attachment possibilities"),
        ("Every student read a book.", "semantic", "one book or multiple books"),
        ("That was impressive.", "pragmatic", "praise or irony"),
    ],
    columns=["Expression", "Ambiguity type", "Possible interpretations"],
)

ambiguity_examples
````

> **Important**
>
> Ambiguity is not necessarily a defect in language. Human communication often
> resolves ambiguity through context, world knowledge, and interaction.

# 10. Linguistic Resources and Annotation

NLP systems may use:

- corpora;
- dictionaries and lexicons;
- morphological analyzers;
- grammars;
- treebanks;
- dependency annotations;
- semantic-role annotations;
- entity and relation annotations;
- WordNet-style lexical databases;
- parallel corpora;
- discourse and coreference datasets.

Annotation converts linguistic analysis into data that models and evaluation
tools can use.

## 10.1 Annotation Layers

One sentence may receive several annotation layers.

```text
Raw text:
The researcher analyzed the document.

Tokens:
[The] [researcher] [analyzed] [the] [document] [.]

Part of speech:
DET NOUN VERB DET NOUN PUNCT

Dependencies:
researcher → subject of analyzed
document   → object of analyzed

Semantic roles:
researcher → agent
document   → affected entity
```

````python
annotated_tokens = pd.DataFrame(
    [
        ("The", "DET", "determiner", None),
        ("researcher", "NOUN", "subject", "agent"),
        ("analyzed", "VERB", "root", "predicate"),
        ("the", "DET", "determiner", None),
        ("document", "NOUN", "object", "affected entity"),
        (".", "PUNCT", "punctuation", None),
    ],
    columns=["Token", "Part of speech", "Dependency", "Semantic role"],
)

annotated_tokens
````

## 10.2 Annotation Is an Analytical Decision

Annotators may disagree because:

- guidelines are unclear;
- categories overlap;
- context is missing;
- the sentence is genuinely ambiguous;
- linguistic theories define structures differently;
- language variation is not represented adequately.

Disagreement should be analyzed rather than automatically treated as careless
annotation.

# 11. Multilingual Considerations

Languages vary in ways that directly affect NLP.

Differences may include:

- writing system and direction;
- word-boundary conventions;
- morphological richness;
- grammatical gender and agreement;
- flexible or fixed word order;
- pro-drop and omitted pronouns;
- compounding;
- clitics;
- politeness systems;
- dialect variation;
- availability of labeled resources.

A pipeline designed for English cannot be assumed to transfer unchanged to
Arabic or another language.

## 11.1 Arabic Examples

Arabic NLP may need to consider:

- attached conjunctions, prepositions, articles, and pronouns;
- root-and-pattern morphology;
- diacritized and undiacritized text;
- Modern Standard Arabic and dialect variation;
- spelling variants;
- right-to-left display;
- rich agreement and inflection.

Tokenization and normalization choices should therefore be documented
specifically for the target Arabic variety and task.

````python
arabic_examples = [
    "والكتاب",
    "وبالكتاب",
    "كتابها",
]

illustrative_segments = {
    "والكتاب": "و + ال + كتاب",
    "وبالكتاب": "و + ب + ال + كتاب",
    "كتابها": "كتاب + ها",
}

for word in arabic_examples:
    print(f"{word} -> {illustrative_segments[word]}")
````

The segmentation above is illustrative. A real morphological analyzer may
produce several analyses depending on context, orthography, and annotation
conventions.

# 12. Linguistic Levels in NLP Systems

| NLP task | Especially relevant linguistic levels |
|---|---|
| Spelling correction | Orthography, morphology, context |
| Sentiment analysis | Lexicon, syntax, pragmatics |
| Named Entity Recognition | Orthography, syntax, semantics |
| Machine translation | Morphology, syntax, semantics, discourse |
| Question answering | Syntax, semantics, pragmatics, knowledge |
| Summarization | Semantics, discourse, information structure |
| Dialogue systems | Pragmatics, discourse, user intent |
| Coreference resolution | Syntax, semantics, discourse |
| Search | Morphology, semantics, user context |

````python
task_levels = {
    "sentiment analysis": {"semantics", "syntax", "pragmatics"},
    "machine translation": {"morphology", "syntax", "semantics", "discourse"},
    "coreference resolution": {"syntax", "semantics", "discourse"},
    "dialogue system": {"semantics", "pragmatics", "discourse"},
}

for task, levels in task_levels.items():
    print(f"{task:<24} -> {', '.join(sorted(levels))}")
````

Modern neural models may learn useful patterns without receiving explicit
linguistic labels. This does not make linguistic analysis irrelevant.
Linguistic concepts remain valuable for:

- task design;
- annotation;
- error analysis;
- evaluation;
- multilingual adaptation;
- interpretability;
- identifying systematic model limitations.
