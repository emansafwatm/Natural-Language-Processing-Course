# 1. What Is Morphology?

**Morphology** is the study of the internal structure of words and the
processes through which words change or are formed.

Consider:

```text
help
helpful
unhelpful
unhelpfulness
```

These forms are related, but each contains different meaningful components.
Morphological analysis identifies those components and the functions they
contribute.

Morphology is important because a single lexical idea may appear in many
surface forms:

```text
connect
connects
connected
connecting
connection
disconnected
reconnection
```

Treating every surface form as unrelated increases vocabulary size and
reduces the amount of evidence available for each underlying word or concept.

````python
from collections import Counter

word_forms = [
    "connect",
    "connects",
    "connected",
    "connecting",
    "connection",
    "disconnected",
    "reconnection",
]

surface_vocabulary = Counter(word_forms)

print("Surface forms:", list(surface_vocabulary))
print("Surface vocabulary size:", len(surface_vocabulary))
````

# 2. Morphemes and Word Structure

A **morpheme** is a minimal unit that contributes lexical meaning or
grammatical function.

Examples:

```text
cat + s
walk + ed
un + kind
teach + er
```

In these examples:

- `cat`, `walk`, `kind`, and `teach` contribute lexical meaning;
- `-s`, `-ed`, `un-`, and `-er` contribute grammatical or derivational
  information.

## 2.1 Free and Bound Morphemes

A **free morpheme** can occur independently as a word:

```text
book
read
kind
```

A **bound morpheme** normally appears only as part of another word:

```text
un-
-ed
-ness
```

Bound morphemes include prefixes, suffixes, infixes, circumfixes, and other
language-specific forms.

## 2.2 Root, Stem, and Base

These terms are related but not identical.

- **Root:** the central lexical element after removable affixes are excluded.
- **Base:** any form to which an affix can be added.
- **Stem:** the form to which an inflectional affix is attached.

Example:

```text
nationalizations

nation        → root or lexical base
national      → base for -ize
nationalize   → base for -ation
nationalization → stem for plural -s
```

Terminology can vary across linguistic traditions, so annotation guidelines
should define the intended analysis.

````python
import pandas as pd

morphological_structure = pd.DataFrame(
    [
        ("cats", "cat + s", "root + plural"),
        ("walked", "walk + ed", "root + past"),
        ("teacher", "teach + er", "root + agent-forming suffix"),
        ("unhelpful", "un + help + ful", "prefix + root + suffix"),
        ("nationalizations", "nation + al + ize + ation + s", "multiple derivational layers + plural"),
    ],
    columns=["Word", "Illustrative segmentation", "Analysis"],
)

morphological_structure
````

# 3. Inflection and Derivation

## 3.1 Inflection

Inflection expresses grammatical information without normally creating a new
dictionary entry.

English examples include:

- number: `cat → cats`;
- tense: `walk → walked`;
- aspect: `walk → walking`;
- comparison: `small → smaller`.

Other languages may mark case, gender, person, mood, definiteness, and many
additional features.

## 3.2 Derivation

Derivation forms a new lexeme or changes meaning or grammatical category.

```text
teach → teacher
happy → happiness
legal → illegal
modern → modernize
```

Derivation is often less predictable than inflection. A theoretically possible
form may not be accepted by speakers or may acquire a specialized meaning.

````python
examples = pd.DataFrame(
    [
        ("book", "books", "inflection", "singular → plural noun"),
        ("write", "wrote", "inflection", "present/base → past"),
        ("teach", "teacher", "derivation", "verb → noun"),
        ("happy", "unhappy", "derivation", "meaning changed"),
        ("modern", "modernize", "derivation", "adjective → verb"),
    ],
    columns=["Base", "Result", "Process", "Effect"],
)

examples
````

> **Important**
>
> Inflection and derivation are linguistic categories, not merely string
> operations. Irregular forms such as `go → went` cannot be recovered by
> deleting a suffix.

# 4. Word-Formation Processes

Languages create words through several processes.

| Process | Description | Example |
|---|---|---|
| Affixation | Add a bound morpheme | `un + happy` |
| Compounding | Combine lexical bases | `data + base` |
| Conversion | Change category without overt affix | `to email` |
| Clipping | Shorten an existing form | `laboratory → lab` |
| Blending | Combine parts of words | `smoke + fog → smog` |
| Acronym formation | Use initial letters or sounds | `NLP`, `NASA` |
| Reduplication | Repeat all or part of a form | language-dependent |
| Borrowing | Adopt a form from another language | widespread across languages |

Word formation affects vocabulary growth, out-of-vocabulary handling, and
semantic interpretation.

````python
word_formation = {
    "database": ("compound", ["data", "base"]),
    "email": ("conversion or noun-to-verb use", ["email"]),
    "lab": ("clipping", ["laboratory"]),
    "smog": ("blend", ["smoke", "fog"]),
    "NLP": ("initialism", ["Natural", "Language", "Processing"]),
}

for word, (process, source) in word_formation.items():
    print(f"{word:<10} | {process:<28} | {source}")
````

# 5. Morphological Typology

Languages vary in how they package grammatical information.

## 5.1 Analytic Languages

Grammatical relationships are often expressed through separate words and word
order.

## 5.2 Agglutinative Languages

Words may contain several affixes whose boundaries and functions are
relatively clear.

## 5.3 Fusional Languages

One affix may express several grammatical features simultaneously.

## 5.4 Polysynthetic Languages

A single word may encode information comparable to an entire sentence in
another language.

## 5.5 Root-and-Pattern Morphology

Semitic languages such as Arabic frequently combine consonantal roots with
patterns and affixes.

Morphological typology is a continuum rather than a perfect classification.
A language may show characteristics of several types.

# 6. Morphology in NLP

Morphological processing supports:

- search and information retrieval;
- vocabulary reduction;
- spelling correction;
- part-of-speech tagging;
- machine translation;
- Named Entity Recognition;
- sentiment analysis;
- language modeling;
- text generation;
- low-resource and multilingual NLP.

Example: a search for `connect` may need to retrieve documents containing
`connected`, `connection`, or `reconnecting`, depending on the application.

````python
documents = [
    "The devices are connected.",
    "Connection quality is stable.",
    "The system is reconnecting.",
    "The cable is disconnected.",
]

literal_matches = [
    document for document in documents
    if "connect" in document.lower()
]

print("Literal substring matches:")
for document in literal_matches:
    print("-", document)
````

A substring rule can provide broad recall, but it is not a linguistic
analyzer. It may overmatch unrelated words or fail when forms are irregular.

# 7. Stemming

**Stemming** applies heuristic rules to remove or replace affixes. The output
is a stem-like form and does not have to be a valid dictionary word.

Example:

```text
studies  → studi
studying → studi
relational → relat
```

Stemming is often fast and useful for search or sparse text models, but it can
produce two important error types:

- **overstemming:** unrelated words are merged;
- **understemming:** related forms remain separate.

````python
def simple_english_stem(word: str) -> str:
    """A deliberately small educational stemmer, not a production tool."""
    word = word.lower()

    replacements = [
        ("ies", "y"),
        ("ing", ""),
        ("ed", ""),
        ("es", ""),
        ("s", ""),
    ]

    for suffix, replacement in replacements:
        if word.endswith(suffix) and len(word) > len(suffix) + 2:
            return word[:-len(suffix)] + replacement

    return word


stemming_examples = [
    "studies",
    "studying",
    "walked",
    "boxes",
    "cats",
    "analysis",
]

for word in stemming_examples:
    print(f"{word:<10} -> {simple_english_stem(word)}")
````

The final example demonstrates why handcrafted suffix deletion is risky:
`analysis` ends in `s`, but the final `s` is not a plural suffix.

# 8. Lemmatization

**Lemmatization** maps an inflected form to a dictionary form called a
**lemma**.

```text
am, is, are → be
went → go
mice → mouse
better → good
```

Accurate lemmatization may require:

- part-of-speech information;
- morphological features;
- a lexicon;
- contextual disambiguation.

Unlike stemming, lemmatization aims to produce a linguistically valid form.

````python
irregular_lemmas = {
    ("went", "VERB"): "go",
    ("was", "VERB"): "be",
    ("mice", "NOUN"): "mouse",
    ("better", "ADJ"): "good",
}

def educational_lemmatize(word: str, part_of_speech: str) -> str:
    key = (word.lower(), part_of_speech.upper())

    if key in irregular_lemmas:
        return irregular_lemmas[key]

    lowered = word.lower()

    if part_of_speech.upper() == "NOUN" and lowered.endswith("s"):
        return lowered[:-1]

    if part_of_speech.upper() == "VERB" and lowered.endswith("ing"):
        return lowered[:-3]

    return lowered


lemma_examples = [
    ("went", "VERB"),
    ("mice", "NOUN"),
    ("cars", "NOUN"),
    ("running", "VERB"),
    ("better", "ADJ"),
]

for word, pos in lemma_examples:
    print(f"{word:<10} {pos:<5} -> {educational_lemmatize(word, pos)}")
````

This is an educational illustration. Production lemmatizers require more
complete language-specific rules, lexicons, or trained models.

## 8.1 Stemming Versus Lemmatization

| Property | Stemming | Lemmatization |
|---|---|---|
| Main method | Heuristic affix removal | Linguistic analysis |
| Output | Stem-like string | Dictionary lemma |
| Context or POS | Often not required | Frequently useful |
| Speed | Usually fast | Often more expensive |
| Linguistic validity | Not guaranteed | Intended |
| Typical use | Retrieval and compact baselines | Linguistically informed pipelines |

# 9. Morphological Segmentation and Analysis

**Segmentation** divides a word into meaningful components.

```text
unhappiness → un + happy + ness
```

**Morphological analysis** may additionally return grammatical features.

```text
cats
lemma: cat
part of speech: noun
number: plural
```

**Morphological generation** performs the reverse operation: it produces a
surface word from a lemma and feature set.

````python
morphological_analysis = {
    "surface": "walked",
    "lemma": "walk",
    "part_of_speech": "VERB",
    "tense": "PAST",
    "segmentation": ["walk", "-ed"],
}

pd.Series(morphological_analysis, name="Illustrative analysis")
````

## 9.1 Finite-State Morphology

Finite-state methods have long been used to represent:

- permitted affix sequences;
- spelling changes;
- mappings between lexical and surface forms;
- morphological analysis and generation.

They are attractive because they are interpretable, efficient, and suitable
for combining lexicons with explicit linguistic rules.

# 10. Morphological Ambiguity

One surface form may have several analyses.

Example:

```text
saw
```

Possible interpretations include:

- past tense of the verb `see`;
- a noun referring to a cutting tool.

Context determines the correct lemma and grammatical analysis.

````python
ambiguous_forms = pd.DataFrame(
    [
        ("saw", "see", "VERB", "past tense"),
        ("saw", "saw", "NOUN", "cutting tool"),
        ("flies", "fly", "NOUN", "plural"),
        ("flies", "fly", "VERB", "third-person singular"),
    ],
    columns=["Surface form", "Lemma", "Part of speech", "Features or meaning"],
)

ambiguous_forms
````

A morphological analyzer may return all valid analyses. A contextual
disambiguation model then selects the most plausible analysis in the sentence.

# 11. Arabic Morphology

Arabic presents important morphological phenomena for NLP:

- consonantal roots and word patterns;
- prefixes and suffixes;
- attached conjunctions and prepositions;
- the definite article;
- attached object and possessive pronouns;
- rich number, gender, person, tense, mood, and voice information;
- diacritized and undiacritized forms;
- spelling variation;
- ambiguity caused by omitted short vowels;
- differences between Modern Standard Arabic and dialects.

## 11.1 Root-and-Pattern Structure

A simplified example uses the root:

```text
ك ت ب
```

It is associated with concepts related to writing. Different patterns and
affixes produce forms such as words conventionally translated as:

- wrote;
- writer;
- book;
- office or desk;
- library.

Root extraction alone does not provide the complete meaning or grammatical
analysis of a word.

````python
arabic_examples = pd.DataFrame(
    [
        ("والكتاب", "و + ال + كتاب", "and + the + book"),
        ("بالمدرسة", "ب + ال + مدرسة", "in/with + the + school"),
        ("كتابها", "كتاب + ها", "her book"),
        ("وسيكتبون", "و + س + يكتب + ون", "and + future + write + plural"),
    ],
    columns=["Surface form", "Illustrative segmentation", "Approximate gloss"],
)

arabic_examples
````

These segmentations are simplified teaching examples. A real Arabic analyzer
may return multiple segmentations and feature structures depending on context,
dialect, tokenization scheme, and annotation conventions.

## 11.2 Arabic Tokenization Choices

Consider:

```text
والكتاب
```

Possible tokenization policies include:

- keep the complete orthographic word;
- split the conjunction;
- split the definite article;
- split all recognized clitics.

Different benchmarks and tools may use different segmentation standards.
Model evaluation is meaningful only when the tokenization policy is explicit
and consistent.

# 12. Evaluation and Error Analysis

Morphological systems may be evaluated on:

- exact lemma accuracy;
- exact segmentation;
- morpheme boundary precision, recall, and F1;
- feature accuracy;
- complete-analysis accuracy;
- out-of-vocabulary performance;
- performance by language variety or morphological category;
- effect on a downstream task.

A system can produce the correct lemma but incorrect grammatical features, so
one aggregate score may hide important errors.

````python
gold_segmentations = {
    "unhelpful": ["un", "help", "ful"],
    "walked": ["walk", "ed"],
    "teacher": ["teach", "er"],
}

predicted_segmentations = {
    "unhelpful": ["un", "helpful"],
    "walked": ["walk", "ed"],
    "teacher": ["teach", "er"],
}

for word in gold_segmentations:
    exact = gold_segmentations[word] == predicted_segmentations[word]
    print(f"{word:<10} exact segmentation correct: {exact}")
````

## 12.1 Common Error Categories

- incorrect lemma;
- missed affix;
- incorrect morpheme boundary;
- overstemming;
- understemming;
- wrong part of speech;
- wrong grammatical feature;
- incorrect analysis of an irregular form;
- failure on an unseen word;
- wrong analysis caused by missing context;
- mismatch between the system and benchmark tokenization scheme.
