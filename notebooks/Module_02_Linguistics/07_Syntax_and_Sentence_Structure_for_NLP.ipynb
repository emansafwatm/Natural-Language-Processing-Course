# 1. What Is Syntax?

**Syntax** studies how words combine into grammatical structures.

Compare:

```text
The student reviewed the article.
The article reviewed the student.
```

The words are similar, but their order and grammatical relations produce
different meanings.

Syntax helps an NLP system identify:

- who performed an action;
- what was affected;
- which words modify other words;
- which expressions form phrases;
- how clauses are connected;
- which interpretations are grammatically possible.

````python
sentence_a = "The student reviewed the article."
sentence_b = "The article reviewed the student."

print(sentence_a.split())
print(sentence_b.split())
print("\nWord order changes the grammatical roles.")
````

# 2. Grammatical Categories

Grammatical categories, often called **parts of speech**, describe how words
behave in sentences.

| Category | Typical function | Examples |
|---|---|---|
| Noun | entity or concept | student, system |
| Verb | action, event, or state | analyze, run, be |
| Adjective | modifies a noun | useful, large |
| Adverb | modifies a verb, adjective, or clause | quickly, very |
| Pronoun | substitutes for a noun phrase | she, it |
| Determiner | introduces a noun phrase | the, a |
| Preposition | expresses a relation | in, with |
| Conjunction | connects units | and, because |
| Auxiliary | supports tense, aspect, voice, or modality | is, have, can |

````python
import pandas as pd

tagged_sentence = pd.DataFrame(
    [
        ("The", "DET"),
        ("researcher", "NOUN"),
        ("carefully", "ADV"),
        ("reviewed", "VERB"),
        ("the", "DET"),
        ("new", "ADJ"),
        ("paper", "NOUN"),
        (".", "PUNCT"),
    ],
    columns=["Token", "Part of speech"],
)

tagged_sentence
````

A word may have different categories in different contexts.

```text
They record the meeting.  → verb
The record is complete.   → noun
```

Part-of-speech tagging therefore requires context.

# 3. Phrases, Clauses, and Sentences

A **phrase** is a group of words functioning as one unit.

- noun phrase: `the new model`;
- verb phrase: `analyzed the document`;
- prepositional phrase: `with a parser`;
- adjective phrase: `very useful`;
- adverb phrase: `extremely quickly`.

A **clause** normally contains a predicate and often a subject.

```text
The model failed.                 → independent clause
because the input was malformed  → dependent clause
```

````python
phrase_examples = pd.DataFrame(
    [
        ("the new model", "noun phrase"),
        ("analyzed the document", "verb phrase"),
        ("with a parser", "prepositional phrase"),
        ("very useful", "adjective phrase"),
        ("extremely quickly", "adverb phrase"),
    ],
    columns=["Expression", "Phrase type"],
)

phrase_examples
````

## 3.1 Sentence Types

- **Simple:** one independent clause.
- **Compound:** coordinated independent clauses.
- **Complex:** an independent clause plus a dependent clause.
- **Compound-complex:** multiple independent clauses plus a dependent clause.

````python
sentence_types = pd.DataFrame(
    [
        ("The parser succeeded.", "simple"),
        ("The parser succeeded, and the system continued.", "compound"),
        ("The system continued because the parser succeeded.", "complex"),
        (
            "The parser succeeded, and the system continued because the input was valid.",
            "compound-complex",
        ),
    ],
    columns=["Sentence", "Type"],
)

sentence_types
````

# 4. Constituency

**Constituency syntax** groups words into nested phrases.

```text
Sentence
├── Noun Phrase
│   ├── Determiner: The
│   └── Noun: researcher
└── Verb Phrase
    ├── Verb: analyzed
    └── Noun Phrase
        ├── Determiner: the
        └── Noun: document
```

A simplified phrase-structure grammar may contain:

```text
S  → NP VP
NP → Det N
VP → V NP
```

````python
grammar = {
    "S": [("NP", "VP")],
    "NP": [("Det", "N"), ("ProperNoun",)],
    "VP": [("V", "NP"), ("V",)],
    "Det": [("the",), ("a",)],
    "N": [("student",), ("article",)],
    "V": [("reviewed",), ("arrived",)],
    "ProperNoun": [("Mona",)],
}

for symbol, productions in grammar.items():
    print(f"{symbol:<11} -> {productions}")
````

A constituency parser may use bracket notation:

```text
(S
  (NP (DET The) (N researcher))
  (VP
    (V analyzed)
    (NP (DET the) (N document))))
```

````python
bracketed_parse = [
    "S",
    ["NP", ["DET", "The"], ["N", "researcher"]],
    [
        "VP",
        ["V", "analyzed"],
        ["NP", ["DET", "the"], ["N", "document"]],
    ],
]

bracketed_parse
````

# 5. Dependency Structure

**Dependency syntax** represents direct relations between words.

```text
analyzed
├── subject → researcher
├── object  → document
└── punctuation → .
```

Each relation contains a head, a dependent, and a label.

````python
dependencies = pd.DataFrame(
    [
        ("analyzed", "researcher", "subject"),
        ("researcher", "The", "determiner"),
        ("analyzed", "document", "object"),
        ("document", "the", "determiner"),
        ("analyzed", ".", "punctuation"),
    ],
    columns=["Head", "Dependent", "Relation"],
)

dependencies
````

| Representation | Main emphasis |
|---|---|
| Constituency | nested phrase structure |
| Dependency | direct grammatical relations among words |

Both can be useful; they emphasize different structural information.

# 6. Grammatical Relations

Important relations include:

- subject;
- direct and indirect object;
- complement;
- modifier;
- determiner;
- auxiliary;
- conjunction;
- oblique or prepositional relation.

In:

> The engineer sent the team a report yesterday.

`engineer` is the subject, `team` is the recipient, `report` is the direct
object, and `yesterday` is a temporal modifier.

````python
grammatical_relations = pd.DataFrame(
    [
        ("engineer", "subject", "sender"),
        ("team", "indirect object", "recipient"),
        ("report", "direct object", "transferred entity"),
        ("yesterday", "temporal modifier", "time"),
    ],
    columns=["Expression", "Syntactic relation", "Approximate semantic role"],
)

grammatical_relations
````

Syntactic relations and semantic roles are related but not identical.

```text
The researcher opened the door.
The door was opened by the researcher.
```

`The researcher` is the semantic agent in both, but its syntactic relation
changes.

# 7. Word Order and Agreement

Commonly discussed basic orders include:

- SVO: subject–verb–object;
- SOV: subject–object–verb;
- VSO: verb–subject–object.

Languages may have relatively fixed or flexible order. Flexible order does not
mean absence of syntax; case, agreement, discourse, and context may carry
structural information.

## 7.1 Agreement

Agreement links grammatical features across words.

```text
The result is correct.
The results are correct.
```

Agreement may involve number, gender, person, case, definiteness, tense, or mood.

````python
agreement_examples = pd.DataFrame(
    [
        ("The model works.", "singular subject + singular verb", True),
        ("The models work.", "plural subject + plural verb", True),
        ("The models works.", "plural subject + singular verb", False),
    ],
    columns=["Sentence", "Pattern", "Standard agreement"],
)

agreement_examples
````

In:

> The results of the experiment are promising.

the verb agrees with `results`, not with the closer singular noun `experiment`.
This is a long-distance dependency.

# 8. Syntactic Ambiguity

A sentence is syntactically ambiguous when more than one grammatical structure is
possible.

## Prepositional-Phrase Attachment

> I saw the researcher with the telescope.

- I used the telescope.
- The researcher had the telescope.

## Coordination Scope

> Old men and women attended.

- both groups were old;
- only the men were old.

````python
ambiguity_cases = pd.DataFrame(
    [
        (
            "I saw the researcher with the telescope.",
            "prepositional-phrase attachment",
            "instrument or researcher modifier",
        ),
        (
            "Old men and women attended.",
            "coordination scope",
            "old modifies one group or both groups",
        ),
        (
            "Visiting researchers can be difficult.",
            "gerund/participle structure",
            "researchers who visit or the act of visiting researchers",
        ),
    ],
    columns=["Sentence", "Ambiguity type", "Interpretations"],
)

ambiguity_cases
````

Context, world knowledge, probabilities, punctuation, and domain conventions may
help resolve ambiguity.

# 9. Parsing in NLP

**Parsing** assigns a structural analysis to a sentence.

Parsing supports:

- information extraction;
- question answering;
- relation extraction;
- grammar checking;
- machine translation;
- semantic-role labeling;
- summarization;
- coreference resolution.

Approaches include handcrafted grammars, chart parsing, probabilistic grammars,
transition-based parsing, graph-based parsing, neural parsing, and
Transformer-based structured prediction.

## 9.1 Dependency Parse Example

A dependency parser predicts a head and relation for each token. A valid tree
normally has one root, one head for each non-root token, connected structure, and
no cycles.

````python
tokens = ["The", "researcher", "analyzed", "the", "document", "."]
heads = [1, 2, -1, 4, 2, 2]
relations = ["det", "nsubj", "root", "det", "obj", "punct"]

dependency_parse = pd.DataFrame(
    {
        "Index": list(range(len(tokens))),
        "Token": tokens,
        "Head index": heads,
        "Relation": relations,
    }
)

dependency_parse
````

````python
for index, token, head_index, relation in zip(
    dependency_parse["Index"],
    dependency_parse["Token"],
    dependency_parse["Head index"],
    dependency_parse["Relation"],
):
    head = "ROOT" if head_index == -1 else tokens[head_index]
    print(f"{index}: {token:<12} -> {head:<12} ({relation})")
````

# 10. Multilingual Syntax

Languages vary in:

- basic word order;
- case marking;
- agreement systems;
- pronoun omission;
- clitic placement;
- question formation;
- negation;
- relative clauses;
- passive and causative constructions.

A parser trained on one language or annotation tradition should not be assumed to
transfer directly to another.

## 10.1 Arabic Syntax

Arabic syntax may involve:

- verbal and nominal sentence patterns;
- VSO and SVO orders;
- rich agreement;
- attached pronouns and clitics;
- omitted subjects recoverable from verb morphology;
- Modern Standard Arabic and dialect variation;
- flexible order influenced by discourse and emphasis.

Simplified example:

```text
كتبَ الطالبُ التقريرَ
wrote the-student the-report
```

````python
arabic_dependency_example = pd.DataFrame(
    [
        ("كتبَ", "ROOT", "root"),
        ("الطالبُ", "كتبَ", "subject"),
        ("التقريرَ", "كتبَ", "object"),
    ],
    columns=["Token", "Head", "Relation"],
)

arabic_dependency_example
````

This is a teaching example. Real Arabic parsing depends on tokenization,
diacritization, variety, annotation scheme, and context.

# 11. Evaluation and Error Analysis

Constituency parsing is often evaluated using labeled or unlabeled precision,
recall, and F1 over predicted constituents.

Dependency parsing commonly uses:

- **UAS:** Unlabeled Attachment Score;
- **LAS:** Labeled Attachment Score.

UAS checks whether a token's predicted head is correct. LAS checks whether both
the head and dependency label are correct.

````python
gold_heads = [1, 2, -1, 4, 2, 2]
predicted_heads = [1, 2, -1, 4, 2, 2]

gold_labels = ["det", "nsubj", "root", "det", "obj", "punct"]
predicted_labels = ["det", "nsubj", "root", "det", "obl", "punct"]

correct_heads = sum(
    gold == predicted
    for gold, predicted in zip(gold_heads, predicted_heads)
)

correct_labeled = sum(
    gold_head == predicted_head and gold_label == predicted_label
    for gold_head, predicted_head, gold_label, predicted_label
    in zip(
        gold_heads,
        predicted_heads,
        gold_labels,
        predicted_labels,
    )
)

uas = correct_heads / len(gold_heads)
las = correct_labeled / len(gold_heads)

print(f"UAS: {uas:.3f}")
print(f"LAS: {las:.3f}")
````

## Common Parsing Errors

- incorrect subject or object attachment;
- incorrect prepositional-phrase attachment;
- coordination errors;
- wrong root;
- incorrect relation label;
- long-distance dependency failure;
- punctuation errors;
- tokenization mismatch;
- domain shift;
- code-mixed or non-standard language.
