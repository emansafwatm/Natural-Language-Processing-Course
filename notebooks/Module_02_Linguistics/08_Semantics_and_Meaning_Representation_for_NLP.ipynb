# 1. What Is Semantics?

**Semantics** studies conventional meaning in words, phrases, and sentences.

Compare:

```text
The student opened the file.
The file opened the student.
```

Both sentences are syntactically possible, but only the first is normally
semantically plausible.

Syntax asks how expressions are structured. Semantics asks what those
expressions mean.

Meaning in NLP may involve:

- lexical meaning;
- relations among words;
- word senses;
- sentence composition;
- semantic roles;
- events and states;
- reference;
- entailment and contradiction;
- similarity;
- structured knowledge.

````python
import pandas as pd

syntax_semantics = pd.DataFrame(
    [
        ("The researcher analyzed the paper.", "well formed", "plausible"),
        ("The paper analyzed the researcher.", "well formed", "usually implausible"),
        ("Researcher the paper analyzed.", "poorly formed in standard English", "difficult to interpret"),
    ],
    columns=["Sentence", "Syntactic status", "Semantic status"],
)

syntax_semantics
````

# 2. Lexical Meaning

**Lexical semantics** studies word meaning.

A lexical item may contribute:

- a concept;
- a category;
- an event;
- a state;
- a relation;
- an evaluative meaning;
- grammatical meaning.

Words do not exist in isolation. Their meanings are connected through
semantic relations and usage patterns.

# 3. Semantic Relations

Important lexical relations include:

| Relation | Description | Example |
|---|---|---|
| Synonymy | similar meaning | begin / start |
| Antonymy | contrasting meaning | hot / cold |
| Hyponymy | subtype relation | dog / animal |
| Hypernymy | supertype relation | animal / dog |
| Meronymy | part–whole relation | wheel / car |
| Holonymy | whole–part relation | car / wheel |
| Polysemy | related senses | head of a person / head of a department |
| Homonymy | unrelated meanings with same form | bank / bank |

````python
semantic_relations = pd.DataFrame(
    [
        ("begin", "start", "synonymy"),
        ("hot", "cold", "antonymy"),
        ("dog", "animal", "hyponym → hypernym"),
        ("wheel", "car", "part → whole"),
        ("head", "department head", "polysemy"),
        ("bank", "river bank", "homonymy or distant polysemy"),
    ],
    columns=["Expression 1", "Expression 2", "Relation"],
)

semantic_relations
````

Semantic relations support:

- query expansion;
- lexical search;
- question answering;
- ontology construction;
- word-sense disambiguation;
- semantic similarity;
- knowledge-graph reasoning.

# 4. Word Senses and Ambiguity

A word may have more than one sense.

```text
bank
```

Possible senses include:

- a financial institution;
- the side of a river.

**Word-sense disambiguation** selects the intended sense in context.

````python
sense_examples = pd.DataFrame(
    [
        ("She deposited money at the bank.", "financial institution"),
        ("They sat on the bank of the river.", "river side"),
        ("The crane lifted the container.", "machine"),
        ("A crane stood near the lake.", "bird"),
    ],
    columns=["Sentence", "Intended sense"],
)

sense_examples
````

Context may provide evidence through:

- neighboring words;
- topic;
- grammatical relations;
- semantic roles;
- domain;
- document context;
- world knowledge.

````python
def simple_bank_sense(sentence: str) -> str:
    """Educational keyword-based sense disambiguation."""
    lowered = sentence.lower()

    financial_terms = {"money", "deposit", "loan", "account", "credit"}
    river_terms = {"river", "water", "shore", "flood"}

    tokens = set(lowered.replace(".", "").split())

    if tokens & financial_terms:
        return "financial institution"
    if tokens & river_terms:
        return "river side"
    return "uncertain"


examples = [
    "The bank approved the loan.",
    "The river overflowed the bank.",
    "The bank is nearby.",
]

for sentence in examples:
    print(f"{sentence:<35} -> {simple_bank_sense(sentence)}")
````

The final sentence remains ambiguous because the rule has insufficient
context. Real systems use broader context and learned representations.

# 5. Compositional Semantics

Compositional semantics studies how the meanings of parts combine into the
meaning of a larger expression.

Compare:

```text
The dog chased the cat.
The cat chased the dog.
```

The words are nearly identical, but grammatical structure changes the event
being described.

````python
from collections import Counter

sentence_1 = "the dog chased the cat"
sentence_2 = "the cat chased the dog"

print(Counter(sentence_1.split()))
print(Counter(sentence_2.split()))
print("\nBag-of-Words is identical, but event roles are reversed.")
````

Compositional meaning depends on:

- lexical meaning;
- syntactic structure;
- scope;
- negation;
- quantification;
- modification;
- context.

Example:

```text
not useful
very useful
useful model
model is useful
```

## 5.1 Scope

Scope determines which expression affects which part of a sentence.

```text
Not every student passed.
```

This normally means that at least one student did not pass. It does not
necessarily mean that no student passed.

Scope is important for negation, quantifiers, modality, and logical
interpretation.

# 6. Semantic Roles and Events

Semantic-role labeling identifies the participants in an event.

In:

> Sara opened the door with a key.

- `Sara` is the agent;
- `the door` is the affected entity;
- `a key` is the instrument.

````python
semantic_roles = pd.DataFrame(
    [
        ("Sara", "agent"),
        ("opened", "predicate"),
        ("the door", "patient / affected entity"),
        ("a key", "instrument"),
    ],
    columns=["Expression", "Semantic role"],
)

semantic_roles
````

Common semantic roles include:

- agent;
- patient;
- experiencer;
- theme;
- instrument;
- source;
- destination;
- location;
- time;
- beneficiary.

Role names and definitions vary among annotation frameworks.

## 6.1 Events and Frames

A semantic frame represents a type of situation and its participants.

Example frame:

```text

```

Frame-based representations support information extraction, question
answering, and knowledge-base construction.

````python
purchase_frame = {
    "frame": "COMMERCIAL_TRANSFER",
    "buyer": "Mona",
    "seller": "the bookstore",
    "goods": "a dictionary",
    "payment": "$25",
    "time": "Monday",
}

pd.Series(purchase_frame, name="Semantic frame")
````

# 7. Entailment, Contradiction, and Presupposition

**Entailment** means that one statement follows from another.

```text
Mona bought a red car.
→ Mona bought a car.
```

**Contradiction** means two statements cannot both be true in the same
interpretation.

```text
The door is open.
The door is closed.
```

**Neutrality** means neither entailment nor contradiction is established.

```text
Mona bought a car.
Mona bought a house.
```

````python
inference_examples = pd.DataFrame(
    [
        ("Mona bought a red car.", "Mona bought a car.", "entailment"),
        ("The door is open.", "The door is closed.", "contradiction"),
        ("Mona bought a car.", "Mona bought a house.", "neutral"),
    ],
    columns=["Premise", "Hypothesis", "Relation"],
)

inference_examples
````

## 7.1 Presupposition

A presupposition is information treated as already established.

```text
Ali stopped smoking.
```

This commonly presupposes that Ali smoked before.

Presuppositions are important in dialogue, information extraction, and
fact-checking because they may be communicated indirectly.

# 8. Symbolic Meaning Representation

Symbolic representations encode meaning using explicit structures.

A simple predicate representation may be:

```text
student(Sara)
document(Paper1)
analyze(Sara, Paper1)
```

Such representations support explicit reasoning and database-style queries.

````python
facts = [
    ("student", "Sara"),
    ("document", "Paper1"),
    ("analyze", "Sara", "Paper1"),
]

for fact in facts:
    print(fact)
````

## 8.1 First-Order Logic Style

A simplified logical form may represent:

> Every student reads a book.

as:

```text
∀x student(x) → ∃y book(y) ∧ reads(x, y)
```

Logical representations make quantification and relations explicit, but
automatic conversion from natural language remains difficult.

## 8.2 Knowledge Graphs

Knowledge graphs represent entities and relations as triples.

```text
(Sara, studies_at, University)
(University, located_in, Cairo)
```

They support structured search, reasoning, recommendation, and question
answering.

````python
knowledge_graph = pd.DataFrame(
    [
        ("Sara", "studies_at", "University"),
        ("University", "located_in", "Cairo"),
        ("Sara", "researches", "NLP"),
    ],
    columns=["Subject", "Relation", "Object"],
)

knowledge_graph
````

# 9. Distributional Semantics

The **distributional hypothesis** states that words used in similar contexts
tend to have related meanings.

This idea motivates vector representations such as:

- count-based context vectors;
- TF-IDF representations;
- Word2Vec;
- GloVe;
- contextual embeddings from Transformer models.

````python
import numpy as np

vectors = {
    "cat": np.array([0.9, 0.8, 0.1]),
    "dog": np.array([0.85, 0.82, 0.12]),
    "airplane": np.array([0.05, 0.1, 0.95]),
}

def cosine_similarity(a: np.ndarray, b: np.ndarray) -> float:
    denominator = np.linalg.norm(a) * np.linalg.norm(b)
    if denominator == 0:
        return 0.0
    return float(np.dot(a, b) / denominator)

pairs = [("cat", "dog"), ("cat", "airplane"), ("dog", "airplane")]

for left, right in pairs:
    score = cosine_similarity(vectors[left], vectors[right])
    print(f"{left:<9} vs {right:<9} -> {score:.3f}")
````

Vector similarity is useful, but similarity does not equal identity or factual
equivalence. Two words may be distributionally close while differing in
polarity, social meaning, or grammatical behavior.

# 10. Semantic Similarity

Semantic similarity estimates how close two meanings are.

It may be applied to:

- words;
- phrases;
- sentences;
- documents;
- queries and passages.

Similarity differs from relatedness.

```text
doctor / physician → similar
doctor / hospital  → related but not synonymous
```

````python
semantic_pairs = pd.DataFrame(
    [
        ("doctor", "physician", "high similarity"),
        ("doctor", "hospital", "related, not synonymous"),
        ("doctor", "mountain", "low similarity"),
    ],
    columns=["Expression 1", "Expression 2", "Interpretation"],
)

semantic_pairs
````

Similarity evaluation may use:

- human ratings;
- correlation with reference judgments;
- retrieval precision;
- clustering quality;
- downstream task performance.

# 11. Semantic Parsing

**Semantic parsing** converts natural language into a structured meaning
representation.

Example:

```text
Which cities are located in Egypt?
```

Possible structured representation:

```text
SELECT city
WHERE located_in(city, Egypt)
```

Semantic parsing supports database queries, virtual assistants, robotics,
question answering, and task-oriented dialogue.

````python
semantic_parse_example = {
    "utterance": "Show all papers written by Sara.",
    "intent": "SEARCH_PUBLICATIONS",
    "slots": {
        "author": "Sara",
        "document_type": "paper",
    },
}

pd.Series(semantic_parse_example, name="Illustrative semantic parse")
````

Semantic parsing challenges include:

- ambiguity;
- missing context;
- unseen expressions;
- complex compositional structure;
- schema mismatch;
- incomplete world knowledge;
- multilingual variation.

# 12. Multilingual Semantics

Challenges include:

- lexical gaps;
- culture-specific concepts;
- different levels of politeness;
- grammaticalized distinctions;
- idioms;
- metaphor;
- ambiguous translations;
- dialect variation.

Arabic NLP may need to account for:

- ambiguity caused by missing short vowels;
- dialect-specific senses;
- root-related words with distinct meanings;
- context-sensitive translation;
- named entities with spelling variation;
- idiomatic expressions;
- semantic differences among Modern Standard Arabic and dialects.

````python
arabic_examples = pd.DataFrame(
    [
        ("علم", "knowledge / science / flag, depending on vocalization and context"),
        ("كتب", "he wrote / books, depending on context"),
        ("عين", "eye / spring / appointed / specific entity, depending on context"),
    ],
    columns=["Surface form", "Possible interpretations"],
)

arabic_examples
````

# 13. Evaluation and Error Analysis

Semantic systems may be evaluated using:

- word-sense accuracy;
- semantic-role precision, recall, and F1;
- entailment accuracy;
- correlation with similarity judgments;
- exact match for logical forms;
- execution accuracy for database queries;
- retrieval metrics;
- downstream task performance.

## Common Semantic Errors

- incorrect word sense;
- synonym confused with relatedness;
- negation ignored;
- wrong semantic role;
- scope error;
- contradiction missed;
- incorrect entity linking;
- world knowledge error;
- literal interpretation of an idiom;
- multilingual sense mismatch;
- plausible but unsupported inference.

````python
error_examples = pd.DataFrame(
    [
        ("The bank approved the loan.", "river side", "wrong word sense"),
        ("Sara did not attend.", "Sara attended", "negation ignored"),
        ("The door was opened by Ali.", "door = agent", "wrong semantic role"),
        ("He kicked the bucket.", "physical kicking event", "idiom interpreted literally"),
    ],
    columns=["Input", "Incorrect interpretation", "Error type"],
)

error_examples
````

Semantic evaluation should use difficult examples involving ambiguity,
negation, quantifiers, lexical variation, implicit information, and domain
knowledge.
