# 1. What Is Pragmatics?

**Pragmatics** studies how language is interpreted in context.

Semantics focuses on conventional meaning. Pragmatics considers:

- speaker intention;
- listener assumptions;
- time and place;
- prior discourse;
- social relationship;
- shared knowledge;
- implied meaning.

Example:

> Can you open the window?

Semantically, the sentence asks about ability. Pragmatically, it is usually a
request.

````python
import pandas as pd

semantic_pragmatic_examples = pd.DataFrame(
    [
        (
            "Can you open the window?",
            "question about ability",
            "request",
        ),
        (
            "It is cold in here.",
            "statement about temperature",
            "possible request to close a window",
        ),
        (
            "That was impressive.",
            "positive evaluation",
            "praise or sarcasm depending on context",
        ),
    ],
    columns=["Utterance", "Literal meaning", "Possible pragmatic meaning"],
)

semantic_pragmatic_examples
````

> **Key Idea**
>
> The same sentence can perform different functions in different situations.

# 2. Context and Interpretation

Context may include:

- **linguistic context:** surrounding words and sentences;
- **situational context:** place, time, objects, and events;
- **social context:** roles and relationships;
- **cultural context:** conventions and shared practices;
- **world knowledge:** facts and expectations;
- **interaction history:** earlier turns and user preferences.

````python
contexts = pd.DataFrame(
    [
        (
            "The room is dark.",
            "A person enters a room.",
            "description",
        ),
        (
            "The room is dark.",
            "A person is standing beside a light switch.",
            "possible request to turn on the light",
        ),
        (
            "The room is dark.",
            "A photographer is preparing a scene.",
            "observation relevant to camera settings",
        ),
    ],
    columns=["Utterance", "Context", "Likely interpretation"],
)

contexts
````

Context does not guarantee one interpretation. It changes the probability and
appropriateness of possible interpretations.

# 3. Speech Acts

A **speech act** is an action performed through language.

Common speech-act categories include:

| Category | Function | Example |
|---|---|---|
| Assertion | present information | The report is complete. |
| Question | request information | Is the report complete? |
| Request | ask someone to act | Please send the report. |
| Command | direct someone to act | Send the report now. |
| Promise | commit to future action | I will send it tomorrow. |
| Apology | acknowledge responsibility | I am sorry for the delay. |
| Thanks | express gratitude | Thank you for your help. |

````python
speech_acts = pd.DataFrame(
    [
        ("The file is ready.", "assertion"),
        ("Is the file ready?", "question"),
        ("Please send the file.", "request"),
        ("I will send the file tomorrow.", "promise"),
        ("I am sorry for the delay.", "apology"),
    ],
    columns=["Utterance", "Speech act"],
)

speech_acts
````

## 3.1 Direct and Indirect Speech Acts

A direct speech act has a grammatical form that closely matches its function.

```text
Close the door. → command
```

An indirect speech act uses one form to perform another function.

```text
Could you close the door? → question form, request function
```

Intent classification must therefore consider more than punctuation or
sentence type.

````python
def simple_speech_act(utterance: str) -> str:
    """Small educational rule set for direct forms only."""
    text = utterance.strip()

    if text.endswith("?"):
        return "question"
    if text.lower().startswith(("please ", "kindly ")):
        return "request"
    if text.lower().startswith(("sorry", "i apologize")):
        return "apology"
    return "statement"


examples = [
    "Is the file ready?",
    "Please send the file.",
    "I am sorry for the delay.",
    "Could you send the file?",
]

for utterance in examples:
    print(f"{utterance:<32} -> {simple_speech_act(utterance)}")
````

The final example exposes the limitation of surface rules: it is grammatically
a question but pragmatically a request.

# 4. Deixis

**Deixis** concerns expressions whose interpretation depends on the context of
speaking.

Main types include:

- person deixis: `I`, `you`, `we`;
- place deixis: `here`, `there`, `this place`;
- time deixis: `now`, `today`, `tomorrow`;
- discourse deixis: `this point`, `the previous section`;
- social deixis: titles and forms of address.

````python
deixis_examples = pd.DataFrame(
    [
        ("I will meet you here tomorrow.", "person, place, and time"),
        ("This is the main issue.", "discourse or situational"),
        ("Professor, could you review this?", "social and person"),
    ],
    columns=["Utterance", "Deictic information"],
)

deixis_examples
````

Without metadata about speaker, listener, time, and location, a system may be
unable to resolve a deictic expression.

# 5. Presupposition

A **presupposition** is information treated as already established.

Examples:

```text
Sara stopped smoking.
```

Commonly presupposes that Sara previously smoked.

```text
The current manager approved the request.
```

Presupposes that there is a contextually identifiable current manager.

````python
presuppositions = pd.DataFrame(
    [
        ("Mona stopped working remotely.", "Mona previously worked remotely."),
        ("Ali returned to Cairo.", "Ali had been in Cairo before."),
        ("The current editor replied.", "There is an identifiable current editor."),
    ],
    columns=["Utterance", "Typical presupposition"],
)

presuppositions
````

Presupposition is important in fact extraction because indirectly conveyed
information should not always be treated as an explicitly asserted fact.

# 6. Implicature

An **implicature** is meaning suggested by an utterance but not stated
directly.

Example:

```text
A: Did all students pass?
B: Some students passed.
```

B's answer often suggests that not all students passed, although the word
`some` is logically compatible with all students passing.

Implicature depends on conversational expectations such as relevance,
informativeness, truthfulness, and clarity.

Consider:

```text
A: Is the report good?
B: The formatting is excellent.
```

The selective praise may imply that the report has other weaknesses.

````python
implicature_examples = pd.DataFrame(
    [
        (
            "Some students passed.",
            "Not all students passed.",
            "scalar implicature",
        ),
        (
            "The formatting is excellent.",
            "Other aspects may be weak.",
            "relevance-based inference",
        ),
        (
            "I have an appointment.",
            "I may be declining the invitation.",
            "context-dependent implication",
        ),
    ],
    columns=["Utterance", "Possible implication", "Type"],
)

implicature_examples
````

Implicatures are defeasible: additional context can cancel them.

```text
Some students passed—in fact, all of them did.
```

# 7. Politeness and Social Meaning

Language expresses social relationships through:

- titles and honorifics;
- indirect requests;
- hedging;
- formality;
- apology;
- gratitude;
- mitigation;
- turn-taking conventions.

Compare:

```text
Send the document.
Please send the document.
Could you send the document when convenient?
```

All may express a similar goal, but their social meanings differ.

````python
politeness_examples = pd.DataFrame(
    [
        ("Send the document.", "direct"),
        ("Please send the document.", "polite direct request"),
        ("Could you send the document?", "indirect request"),
        (
            "Could you send the document when convenient?",
            "indirect and mitigated request",
        ),
    ],
    columns=["Utterance", "Pragmatic characterization"],
)

politeness_examples
````

Politeness strategies vary across languages, cultures, institutions, and
communities. One universal scoring rule is unlikely to be reliable.

# 8. Dialogue Acts and Conversation State

A **dialogue act** describes the function of one conversational turn.

Examples include:

- greeting;
- request;
- confirmation;
- clarification;
- answer;
- correction;
- acceptance;
- rejection;
- closing.

````python
dialogue = pd.DataFrame(
    [
        (1, "User", "I need to reset my password.", "request"),
        (2, "System", "Do you still have access to your email?", "clarification question"),
        (3, "User", "Yes.", "confirmation"),
        (4, "System", "I will send a reset link.", "inform / commitment"),
    ],
    columns=["Turn", "Speaker", "Utterance", "Dialogue act"],
)

dialogue
````

A dialogue system may maintain a structured state:

```text
intent: reset_password
email_access: yes
authentication_status: pending
next_action: send_reset_link
```

````python
dialogue_state = {
    "intent": "reset_password",
    "email_access": True,
    "authentication_status": "pending",
    "next_action": "send_reset_link",
}

pd.Series(dialogue_state, name="Dialogue state")
````

Dialogue state helps a system avoid asking the same question repeatedly and
supports consistent multi-turn behavior.

# 9. Discourse and Coherence

**Discourse** is language extending beyond one isolated sentence.

A discourse is coherent when its parts form a meaningful whole.

Example:

```text
The server failed during the update.
As a result, the application became unavailable.
Engineers restored service after thirty minutes.
```

The sentences are linked through cause, result, and temporal progression.

## 9.1 Cohesion and Coherence

**Cohesion** concerns observable linguistic links, such as:

- pronouns;
- conjunctions;
- repeated words;
- lexical chains;
- substitutions;
- ellipsis.

**Coherence** concerns the overall conceptual and logical connection among
discourse units.

A text may contain cohesive markers but still be conceptually incoherent.

````python
discourse_features = pd.DataFrame(
    [
        ("The server failed. It was restarted.", "pronoun reference", "cohesion"),
        ("The server failed. Therefore, service stopped.", "causal connective", "cohesion and coherence"),
        ("The server failed. Bananas are yellow.", "no meaningful relation", "incoherent"),
    ],
    columns=["Text", "Observed relation", "Interpretation"],
)

discourse_features
````

# 10. Coreference and Reference Resolution

**Coreference** occurs when two or more expressions refer to the same entity.

```text
Mona submitted the paper.
She received confirmation.
```

`Mona` and `She` refer to the same person.

Related concepts include:

- **anaphora:** reference to an earlier expression;
- **cataphora:** reference to a later expression;
- **exophora:** reference to something in the external situation.

Examples:

```text
Mona arrived. She sat down.        → anaphora
Before she spoke, Mona paused.     → cataphora
Put it over there.                 → exophora
```

````python
reference_examples = pd.DataFrame(
    [
        ("Mona arrived. She sat down.", "She → Mona", "anaphora"),
        ("Before she spoke, Mona paused.", "she → Mona", "cataphora"),
        ("Put it over there.", "depends on external context", "exophora"),
    ],
    columns=["Text", "Reference", "Type"],
)

reference_examples
````

Coreference resolution may use:

- number and gender agreement;
- grammatical role;
- semantic compatibility;
- recency;
- discourse prominence;
- world knowledge;
- document-level context.

````python
def simple_pronoun_candidates(entities, pronoun_features):
    """Filter entities using simple agreement features."""
    matches = []

    for entity in entities:
        number_ok = entity["number"] == pronoun_features["number"]
        gender_ok = (
            pronoun_features["gender"] == "unknown"
            or entity["gender"] == pronoun_features["gender"]
        )

        if number_ok and gender_ok:
            matches.append(entity["name"])

    return matches


entities = [
    {"name": "Mona", "number": "singular", "gender": "feminine"},
    {"name": "the papers", "number": "plural", "gender": "unknown"},
]

pronoun = {"number": "singular", "gender": "feminine"}

simple_pronoun_candidates(entities, pronoun)
````

Agreement filters candidates but does not always resolve the reference. Several
compatible entities may remain.

# 11. Discourse Relations

Discourse relations connect clauses and sentences.

Common relations include:

| Relation | Example marker |
|---|---|
| Cause | because |
| Result | therefore |
| Contrast | however |
| Condition | if |
| Elaboration | for example |
| Sequence | then |
| Concession | although |
| Purpose | so that |

````python
discourse_relations = pd.DataFrame(
    [
        ("The input was invalid, so the request failed.", "result"),
        ("Although the model was small, it performed well.", "concession"),
        ("The system paused because approval was missing.", "cause"),
        ("First collect the data, then validate it.", "sequence"),
    ],
    columns=["Sentence", "Relation"],
)

discourse_relations
````

Connectives provide useful evidence but are not always explicit.

```text
The input was invalid. The request failed.
```

A causal relation may be inferred even without the word `because`.

# 12. Context in NLP Systems

Context-aware NLP appears in:

- dialogue systems;
- machine translation;
- summarization;
- coreference resolution;
- question answering;
- sentiment analysis;
- information extraction;
- Large Language Models.

Systems may use:

- a local sentence window;
- a complete document;
- conversation history;
- retrieved external documents;
- structured dialogue state;
- user-approved profile information;
- tool outputs.

## 12.1 Context Windows

A model context window limits how much input can be processed at one time.

Larger context does not automatically guarantee correct use of information.
A system may:

- ignore relevant earlier evidence;
- focus on distracting details;
- confuse speakers or entities;
- contradict previous turns;
- use stale information;
- exceed privacy boundaries.

````python
conversation = [
    "User: My preferred delivery day is Thursday.",
    "System: Noted.",
    "User: Please schedule the package for my preferred day.",
]

print("\n".join(conversation))
print("\nRequired inference: preferred day = Thursday.")
````

## 12.2 Memory Versus Immediate Context

Immediate context contains information supplied in the current input.

Persistent memory stores information across sessions or interactions.
Persistent memory raises additional requirements involving:

- user consent;
- correction and deletion;
- data minimization;
- access control;
- expiration;
- relevance;
- protection against false assumptions.

# 14. Evaluation and Error Analysis

Pragmatic and discourse systems may be evaluated using:

- dialogue-act accuracy or macro F1;
- intent classification metrics;
- coreference precision, recall, and F1;
- entity-level coreference metrics;
- discourse-relation accuracy;
- response relevance;
- consistency across turns;
- task completion rate;
- human judgments of appropriateness.

````python
gold_links = {
    ("Mona", "She"),
    ("the paper", "it"),
    ("the researchers", "they"),
}

predicted_links = {
    ("Mona", "She"),
    ("the report", "it"),
    ("the researchers", "they"),
}

true_positive = len(gold_links & predicted_links)
precision = true_positive / len(predicted_links)
recall = true_positive / len(gold_links)
f1 = 2 * precision * recall / (precision + recall)

print(f"Precision: {precision:.3f}")
print(f"Recall:    {recall:.3f}")
print(f"F1:        {f1:.3f}")
````

## Common Error Categories

- direct form confused with intended speech act;
- deixis unresolved;
- presupposition treated as explicit assertion;
- implicature missed or invented;
- sarcasm interpreted literally;
- wrong dialogue state;
- pronoun linked to the wrong entity;
- discourse relation misclassified;
- contradiction with earlier context;
- culturally inappropriate politeness;
- context from the wrong user or session;
- irrelevant retrieved context.

````python
error_examples = pd.DataFrame(
    [
        (
            "Could you close the door?",
            "classified only as a factual question",
            "indirect speech act missed",
        ),
        (
            "Mona spoke to Sara after she arrived.",
            "she → wrong person",
            "coreference ambiguity",
        ),
        (
            "Great job!",
            "always positive",
            "sarcasm/context ignored",
        ),
        (
            "Use my preferred date.",
            "asks again despite prior answer",
            "dialogue state lost",
        ),
    ],
    columns=["Input", "Incorrect behavior", "Error type"],
)

error_examples
````

Evaluation should include multi-turn examples, ambiguous references, indirect
language, context shifts, unsupported assumptions, and culturally varied
expressions.
