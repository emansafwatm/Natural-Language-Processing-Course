# 1. What Is a Linguistic Resource?

A **linguistic resource** is a structured source of language data or
linguistic knowledge used for research, training, evaluation, or deployment.

Common resources include:

- raw text corpora;
- speech corpora;
- parallel corpora;
- dictionaries and lexicons;
- terminology databases;
- ontologies and knowledge graphs;
- treebanks;
- morphological databases;
- part-of-speech annotations;
- Named Entity Recognition datasets;
- semantic-role and coreference corpora;
- dialogue and discourse datasets.

````python
import pandas as pd

resource_types = pd.DataFrame(
    [
        ("Raw corpus", "Unannotated language data", "language modeling"),
        ("Parallel corpus", "Aligned text in two or more languages", "machine translation"),
        ("Lexicon", "Words with linguistic information", "morphology or sentiment"),
        ("Treebank", "Sentences with syntactic annotation", "parsing"),
        ("Ontology", "Concepts and formal relations", "knowledge-based NLP"),
        ("Dialogue corpus", "Multi-turn interactions", "conversation systems"),
    ],
    columns=["Resource", "Content", "Typical use"],
)

resource_types
````

> **Key Idea**
>
> A model learns only from the information represented in its data and
> annotation scheme.

# 2. Corpora and Corpus Types

A **corpus** is a structured collection of language data.

Corpora may differ by:

- language;
- domain;
- medium;
- time period;
- speaker or author population;
- annotation;
- size;
- licensing;
- collection method.

## 2.1 Common Corpus Categories

| Corpus type | Description |
|---|---|
| General corpus | Broad language use across several domains |
| Specialized corpus | One domain, profession, or genre |
| Comparable corpus | Similar content in different languages without direct alignment |
| Parallel corpus | Source and translation pairs |
| Learner corpus | Language produced by learners |
| Historical corpus | Language from an earlier period |
| Spoken corpus | Transcribed speech and possibly audio |
| Dialogue corpus | Multi-turn interaction |
| Monitor corpus | Continuously updated language data |

````python
corpus_examples = pd.DataFrame(
    [
        ("News articles from several publishers", "general or news corpus"),
        ("Medical discharge summaries", "specialized corpus"),
        ("Arabic-English translated sentences", "parallel corpus"),
        ("Student essays with error labels", "learner corpus"),
        ("Customer-service conversations", "dialogue corpus"),
    ],
    columns=["Collection", "Corpus type"],
)

corpus_examples
````

# 3. Corpus Design and Sampling

Corpus design should begin with the intended use.

Important questions include:

- Which population should the corpus represent?
- Which languages, dialects, and scripts are required?
- Which domains and genres are relevant?
- Which time period should be covered?
- How will records be sampled?
- Which rare but important cases must be included?
- Which metadata should be retained?
- Which privacy and licensing restrictions apply?

## 3.1 Representativeness

A corpus is representative only relative to a clearly defined target.

A dataset of formal news articles is not representative of:

- social-media language;
- spoken conversation;
- medical notes;
- dialectal Arabic;
- children's language;
- customer-support messages.

Size cannot repair a systematic sampling mismatch.

````python
sample_distribution = pd.DataFrame(
    [
        ("News", 600),
        ("Web articles", 250),
        ("Conversation", 100),
        ("Social media", 50),
    ],
    columns=["Genre", "Documents"],
)

sample_distribution["Percentage"] = (
    sample_distribution["Documents"]
    / sample_distribution["Documents"].sum()
    * 100
).round(1)

sample_distribution
````

## 3.2 Metadata

Useful metadata may include:

- date;
- source;
- language and variety;
- genre;
- channel;
- speaker or author role;
- document identifier;
- collection method;
- annotation version.

Metadata supports filtering and error analysis, but sensitive fields should be
minimized and protected.

# 4. Lexicons, Ontologies, and Knowledge Bases

A **lexicon** stores information about lexical items.

It may contain:

- lemma;
- part of speech;
- pronunciation;
- morphological features;
- senses;
- sentiment;
- frequency;
- translations;
- domain labels.

````python
lexicon = pd.DataFrame(
    [
        ("run", "VERB", "run", "movement or operation"),
        ("running", "VERB", "run", "progressive form"),
        ("bank", "NOUN", "bank", "financial institution"),
        ("bank", "NOUN", "bank", "river side"),
    ],
    columns=["Surface form", "Part of speech", "Lemma", "Sense"],
)

lexicon
````

An **ontology** represents concepts and formal relationships.

Example:

```text
dog       is_a animal
animal    is_a living_thing
wheel     part_of car
Cairo     located_in Egypt
```

Ontologies and knowledge graphs support structured search, reasoning, entity
linking, and question answering.

````python
knowledge_relations = pd.DataFrame(
    [
        ("dog", "is_a", "animal"),
        ("animal", "is_a", "living_thing"),
        ("wheel", "part_of", "car"),
        ("Cairo", "located_in", "Egypt"),
    ],
    columns=["Subject", "Relation", "Object"],
)

knowledge_relations
````

# 5. Treebanks and Annotated Corpora

A **treebank** is a corpus annotated with syntactic structure.

It may contain:

- tokens;
- part-of-speech tags;
- morphological features;
- constituency trees;
- dependency heads;
- dependency labels.

Other annotated corpora may contain entities, relations, events, semantic
roles, coreference, sentiment, dialogue acts, or discourse relations.

````python
annotated_sentence = pd.DataFrame(
    [
        ("The", "DET", 1, "det"),
        ("researcher", "NOUN", 2, "nsubj"),
        ("analyzed", "VERB", -1, "root"),
        ("the", "DET", 4, "det"),
        ("paper", "NOUN", 2, "obj"),
        (".", "PUNCT", 2, "punct"),
    ],
    columns=["Token", "POS", "Head index", "Dependency"],
)

annotated_sentence
````

Annotation reflects a theory and a set of conventions. Two treebanks may
analyze the same sentence differently while remaining internally consistent.

# 6. Annotation Schemes

An **annotation scheme** defines:

- which units are labeled;
- which labels are allowed;
- how boundaries are selected;
- how ambiguous cases are handled;
- whether multiple labels are permitted;
- which context is available;
- how uncertainty is recorded.

## 6.1 Common Annotation Layers

| Layer | Example output |
|---|---|
| Tokenization | `[The] [model] [works]` |
| Part of speech | `DET NOUN VERB` |
| Morphology | number=singular, tense=present |
| Named entities | `Cairo → LOCATION` |
| Sentiment | positive, negative, neutral |
| Semantic roles | agent, patient, instrument |
| Coreference | `Mona ↔ she` |
| Dialogue act | request, answer, confirmation |
| Discourse relation | cause, contrast, result |

````python
annotation_layers = pd.DataFrame(
    [
        ("Cairo", "PROPN", "LOCATION", None),
        ("hosts", "VERB", None, "predicate"),
        ("the conference", "NOUN PHRASE", None, "theme"),
    ],
    columns=["Expression", "POS or phrase", "Entity", "Semantic role"],
)

annotation_layers
````

# 7. Annotation Guidelines

Annotation guidelines convert an abstract task into repeatable decisions.

Strong guidelines include:

- clear label definitions;
- inclusion and exclusion rules;
- positive examples;
- negative examples;
- boundary rules;
- ambiguous examples;
- priority rules;
- escalation procedures;
- version history.

## 7.1 Example: Support Intent Labels

- `account`: login, password, profile, verification, or security;
- `billing`: payment, invoice, charge, refund, or subscription;
- `technical`: application errors, performance, integrations, or failures;
- `delivery`: shipping, courier, package, tracking, or address.

Ambiguous example:

> I cannot access the invoice page.

Possible labels:

- `account` if access is blocked by authentication;
- `billing` if the issue is invoice retrieval;
- `technical` if the page is malfunctioning.

The guideline must state which evidence receives priority.

````python
guidelines = {
    "account": "Credentials, profile, access, verification, and security",
    "billing": "Payments, charges, invoices, refunds, and plans",
    "technical": "Software behavior, errors, and integrations",
    "delivery": "Shipping, tracking, courier, and package status",
}

pd.Series(guidelines, name="Guideline")
````

# 8. Annotation Workflow

A practical workflow may include:

```text
Define task
    ↓
Draft scheme and guidelines
    ↓
Train annotators
    ↓
Pilot annotation
    ↓
Measure disagreement
    ↓
Revise guidelines
    ↓
Annotate full dataset
    ↓
Quality checks and adjudication
    ↓
Freeze and version release
```

## 8.1 Pilot Annotation

A pilot should include:

- common examples;
- rare categories;
- ambiguous cases;
- difficult boundaries;
- multilingual variation;
- expected deployment conditions.

Pilot disagreements often reveal problems in the task definition before they
affect the entire dataset.

# 9. Inter-Annotator Agreement

Agreement measures whether annotators apply the scheme consistently.

Raw agreement is:

\[
	ext{Agreement} =
rac{	ext{matching labels}}{	ext{total items}}
\]

Raw agreement does not account for agreement expected by chance.

````python
annotator_a = ["billing", "technical", "account", "delivery", "billing"]
annotator_b = ["billing", "technical", "technical", "delivery", "billing"]

matches = sum(
    left == right
    for left, right in zip(annotator_a, annotator_b)
)

raw_agreement = matches / len(annotator_a)

print(f"Matching labels: {matches}")
print(f"Raw agreement: {raw_agreement:.3f}")
````

## 9.1 Cohen's Kappa

Cohen's kappa adjusts observed agreement using expected chance agreement.

\[
\kappa =
rac{p_o - p_e}{1 - p_e}
\]

where:

- \(p_o\) is observed agreement;
- \(p_e\) is expected agreement by chance.

````python
from sklearn.metrics import cohen_kappa_score

kappa = cohen_kappa_score(annotator_a, annotator_b)

print(f"Cohen's kappa: {kappa:.3f}")
````

Kappa should be interpreted with the label distribution, task difficulty,
number of annotators, and type of disagreement. One universal threshold is
not appropriate for every annotation problem.

## 9.2 Span Agreement

Sequence tasks require agreement on both boundaries and labels.

Example:

```text
New York University
```

One annotator may label the entire expression as one organization. Another may
select only `New York`.

````python
gold_spans = {
    (0, 3, "ORG"),
    (5, 6, "PERSON"),
    (8, 9, "LOCATION"),
}

predicted_spans = {
    (0, 2, "ORG"),
    (5, 6, "PERSON"),
    (8, 9, "LOCATION"),
}

true_positive = len(gold_spans & predicted_spans)
precision = true_positive / len(predicted_spans)
recall = true_positive / len(gold_spans)
f1 = 2 * precision * recall / (precision + recall)

print(f"Exact-span precision: {precision:.3f}")
print(f"Exact-span recall:    {recall:.3f}")
print(f"Exact-span F1:        {f1:.3f}")
````

# 10. Quality Control and Adjudication

Quality control may include:

- duplicate examples;
- hidden gold-standard items;
- random review;
- automated consistency checks;
- disagreement reports;
- annotator feedback;
- boundary validation;
- label-distribution monitoring;
- adjudication by a specialist.

**Adjudication** resolves disagreements and creates a final reference label.

Adjudication should also improve the guidelines. Otherwise, the same ambiguity
will repeatedly affect later batches.

````python
annotation_batch = pd.DataFrame(
    [
        ("T001", "billing", "billing", "billing"),
        ("T002", "technical", "account", "technical"),
        ("T003", "delivery", "delivery", "delivery"),
        ("T004", "account", "technical", "account"),
    ],
    columns=["Item", "Annotator A", "Annotator B", "Adjudicated label"],
)

annotation_batch["Initial agreement"] = (
    annotation_batch["Annotator A"]
    == annotation_batch["Annotator B"]
)

annotation_batch
````

# 11. Dataset Splits and Leakage

Annotated data is commonly divided into:

- training data;
- validation data;
- test data.

Leakage occurs when test information affects training or model selection.

Examples:

- duplicate documents across splits;
- messages from one conversation divided across splits;
- preprocessing fitted before splitting;
- future examples used to predict the past;
- annotator comments or label names exposed as features.

````python
documents = pd.DataFrame(
    [
        ("D1", "Conversation A", "reset password"),
        ("D2", "Conversation A", "email code missing"),
        ("D3", "Conversation B", "invoice request"),
        ("D4", "Conversation C", "package delayed"),
    ],
    columns=["Document", "Group", "Text"],
)

documents
````

If documents from `Conversation A` are divided between training and test
partitions, evaluation may benefit from repeated conversational context.
Group-aware splitting is safer.

# 12. Documentation, Licensing, and Privacy

A dataset release should document:

- motivation;
- creators and maintainers;
- data sources;
- collection dates;
- sampling method;
- annotation scheme;
- annotator population;
- agreement results;
- preprocessing;
- language and domain coverage;
- known limitations;
- intended and prohibited uses;
- license;
- access restrictions;
- privacy protections.

## 12.1 Datasheets and Data Statements

Dataset documentation may use:

- datasheets for datasets;
- data statements;
- dataset cards;
- corpus-specific documentation.

These formats make assumptions and limitations visible to later users.

## 12.2 Licensing

A dataset license may regulate:

- redistribution;
- commercial use;
- modification;
- attribution;
- derivative datasets;
- model training;
- access to sensitive data.

Public availability does not automatically mean unrestricted use.

## 12.3 Privacy

Language data may contain:

- names;
- email addresses;
- phone numbers;
- locations;
- health information;
- financial information;
- confidential conversations;
- identifying writing patterns.

Privacy protection may require minimization, redaction, access control,
retention limits, secure storage, and review of memorization risks.

````python
import re

email_pattern = re.compile(
    r"\b[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}\b"
)
phone_pattern = re.compile(
    r"\b(?:\+?\d[\d -]{7,}\d)\b"
)

text = "Contact sara@example.com or call +20 10 1234 5678."

redacted = email_pattern.sub("EMAIL_ADDRESS", text)
redacted = phone_pattern.sub("PHONE_NUMBER", redacted)

print(redacted)
````

Automatic redaction is imperfect. Sensitive-data handling should include
human review and documented risk controls where required.

# 13. Multilingual and Arabic Resources

Multilingual resource creation must consider:

- dialects and regional varieties;
- script and orthography;
- tokenization;
- code-switching;
- translation equivalence;
- cultural concepts;
- annotator expertise;
- unequal resource availability.

## 13.1 Low-Resource NLP

A language or variety may be low-resource because it lacks:

- large corpora;
- standardized spelling;
- annotated datasets;
- tokenizers;
- lexicons;
- pretrained models;
- evaluation benchmarks;
- funded annotation efforts.

Low-resource does not mean linguistically simple or socially unimportant.

## 13.2 Arabic Resource Design

Arabic resources may need to specify:

- Modern Standard Arabic or dialect;
- diacritized or undiacritized text;
- segmentation standard;
- treatment of clitics;
- spelling normalization;
- Arabizi or Latin-script Arabic;
- code-switching;
- regional coverage;
- annotation of gender, number, case, mood, and definiteness.

````python
arabic_resource_decisions = pd.DataFrame(
    [
        ("Language variety", "Modern Standard Arabic and Egyptian Arabic"),
        ("Script", "Arabic script; Arabizi excluded"),
        ("Diacritics", "Preserved when present"),
        ("Tokenization", "Clitics segmented according to documented scheme"),
        ("Code-switching", "Arabic-English examples retained and labeled"),
    ],
    columns=["Design decision", "Example specification"],
)

arabic_resource_decisions
````

Consistency is more important than silently mixing incompatible conventions.
